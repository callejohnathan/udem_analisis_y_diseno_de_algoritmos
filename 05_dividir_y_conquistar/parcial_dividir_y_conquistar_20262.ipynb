## Punto 1 — Consultas de presupuesto en una fila de tiendas

En una calle hay `n` tiendas puestas en fila. Cada tienda vende una única botella de una bebida especial, a un precio fijo. Durante `q` días distintos, un comprador llega con una cantidad de dinero diferente cada día. Para cada día, el comprador quiere saber en cuántas tiendas de la calle podría pagar la botella, es decir, cuántas tiendas tienen un precio que no supera el dinero que trae ese día.

### Entrada

* Un entero `n`: número de tiendas.
* Una lista de `n` enteros `precios`, donde `precios[i]` es el precio de la botella en la tienda `i`.
* Un entero `q`: número de consultas (días).
* Una lista de `q` enteros `consultas`, donde `consultas[j]` es el dinero disponible el día `j`.

### Salida

* Una lista de `q` enteros, donde el elemento en la posición `j` es la cantidad de tiendas cuyo precio no supera `consultas[j]`.

### Restricciones

* 1 ≤ n, q ≤ 100000
* 1 ≤ precios[i] ≤ 100000
* 1 ≤ consultas[j] ≤ 10^9

### Ejemplo

`precios = [5, 3, 8, 3, 9]`, `consultas = [1, 3, 7, 9, 20]`

Salida: `[0, 2, 3, 5, 5]`

Explicación: con 1 moneda no alcanza ninguna tienda; con 3 monedas alcanzan las dos tiendas de precio 3; con 7 monedas alcanzan las de precio 3, 3 y 5; con 9 o más monedas alcanzan las 5 tiendas.

```python
casos_ej_033 = [
    {"entrada": {"precios": [5, 3, 8, 3, 9], "consultas": [1, 3, 7, 9, 20]}, "salida": [0, 2, 3, 5, 5]},
    {"entrada": {"precios": [100000], "consultas": [1, 99999, 100000, 1000000000]}, "salida": [0, 0, 1, 1]},
    {"entrada": {"precios": [1], "consultas": [1]}, "salida": [1]},
    {"entrada": {"precios": [2, 2, 2, 2], "consultas": [1, 2, 3]}, "salida": [0, 4, 4]},
]
```

## Punto 2 — El k-ésimo mayor valor

Tienes una lista de enteros sin ningún orden particular. Quieres determinar cuál sería el valor que queda en la posición `k` si la lista se ordenara completa de mayor a menor. Si hay valores repetidos, cada repetición ocupa su propia posición (no se trata de valores distintos).

### Entrada

* Una lista de enteros `nums`.
* Un entero `k`.

### Salida

* El entero que ocupa la posición `k` al ordenar `nums` de mayor a menor.

### Restricciones

* 1 ≤ k ≤ len(nums) ≤ 10^5
* -10^4 ≤ nums[i] ≤ 10^4

### Ejemplo

`nums = [3, 2, 3, 1, 2, 4, 5, 5, 6]`, `k = 4`

Salida: `4`

Explicación: ordenada de mayor a menor la lista queda `[6, 5, 5, 4, 3, 3, 2, 2, 1]`; la posición 4 es el valor `4`.

```python
casos_ej_034 = [
    {"entrada": {"nums": [3, 2, 1, 5, 6, 4], "k": 2}, "salida": 5},
    {"entrada": {"nums": [3, 2, 3, 1, 2, 4, 5, 5, 6], "k": 4}, "salida": 4},
    {"entrada": {"nums": [7], "k": 1}, "salida": 7},
    {"entrada": {"nums": [1, 1, 1, 1], "k": 2}, "salida": 1},
    {"entrada": {"nums": [9, 8, 7, 6, 5, 4, 3, 2, 1], "k": 1}, "salida": 9},
]
```
