## Punto 1 — Subsecuencia más larga con suma limitada

Se tiene una lista de enteros `nums` de longitud `n` y una lista de enteros `consultas` de longitud `m`.

Para cada consulta `consultas[i]`, determine el **tamaño máximo** de una subsecuencia de `nums` cuya suma de elementos sea **menor o igual** que `consultas[i]`.

Una **subsecuencia** es una lista que se obtiene de otra eliminando algunos elementos, o ninguno, sin cambiar el orden de los elementos que quedan.

**Entrada:** las listas `nums` y `consultas`.

**Salida:** una lista `respuesta` de longitud `m`, donde `respuesta[i]` es el tamaño máximo de una subsecuencia de `nums` con suma menor o igual que `consultas[i]`.

**Restricciones:**
- `1 ≤ n, m ≤ 1000`
- `1 ≤ nums[i], consultas[i] ≤ 10^6`
- La solución debe tener una complejidad temporal de **O((n + m) · log n)**.

**Ejemplo:** con `nums = [4, 5, 2, 1]` y `consultas = [3, 10, 21]`:
- La subsecuencia `[2, 1]` suma `3`, y no hay una de tamaño 3 que sume como máximo 3. La respuesta es `2`.
- La subsecuencia `[4, 5, 1]` suma `10`, y no hay una de tamaño 4 que sume como máximo 10. La respuesta es `3`.
- La subsecuencia `[4, 5, 2, 1]` suma `12 ≤ 21`. La respuesta es `4`.

La salida es `[2, 3, 4]`.
