**Contexto.** Un equipo de desarrollo está construyendo una herramienta de revisión de código, parecida a la que muestran los sistemas de control de versiones cuando se comparan dos versiones de un archivo. La herramienta trata cada archivo como una lista de líneas y solo sabe hacer dos operaciones:

- **borrar** una línea de la versión vieja (se muestra con `-`),
- **insertar** una línea de la versión nueva (se muestra con `+`).

Una línea solo se conserva si es idéntica carácter a carácter en ambas versiones, y las líneas conservadas deben mantener su orden relativo. No existe la operación "modificar": cambiar una línea cuenta como un borrado más una inserción.

El equipo necesita el **número mínimo de operaciones** para transformar la versión vieja en la nueva, porque ese número es el que se reporta en el resumen de cada revisión.

Por ejemplo:

```
 Versión vieja        Versión nueva          Diferencia mostrada
 ─────────────        ─────────────          ───────────────────

 x = 1                x = 1                    x = 1
 y = 2                z = 3                  - y = 2
 print(x)             print(x)               + z = 3
 print(y)             print(z)                 print(x)
                                             - print(y)
                                             + print(z)
                                                        → 4 operaciones
```

**Entrada:** dos listas de cadenas, `version_vieja` y `version_nueva`, donde cada cadena es una línea del archivo.

**Salida:** el número mínimo de operaciones (borrados más inserciones) para transformar `version_vieja` en `version_nueva`.

**Restricciones:** cada versión tiene entre 0 y 1000 líneas.

**Entrada para el árbol de la Parte A.3:** `version_vieja = ["A", "B", "C"]`, `version_nueva = ["B", "C", "D"]`, donde cada letra representa una línea completa.
