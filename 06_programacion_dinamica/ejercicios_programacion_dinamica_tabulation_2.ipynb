# Ejercicio 1

---

#### 📘 Descripción del problema

Dado un número entero `n`, escribe un programa que **retorne un arreglo `ans` de longitud `n + 1`**, donde cada posición `i` (0 ≤ i ≤ n) contenga el **número de bits en 1** en la representación binaria de `i`.

En otras palabras, `ans[i]` debe indicar cuántos **unos** tiene el número `i` cuando se expresa en binario.

---

#### 🧠 Ejemplo 1

**Entrada:**
`n = 2`

**Salida:**
`[0, 1, 1]`

**Explicación:**

```
0 --> 0
1 --> 1
2 --> 10
```

---

#### 🧠 Ejemplo 2

**Entrada:**
`n = 5`

**Salida:**
`[0, 1, 1, 2, 1, 2]`

**Explicación:**

```
0 --> 0
1 --> 1
2 --> 10
3 --> 11
4 --> 100
5 --> 101
```

---

#### ⚙️ Restricciones

* 0 ≤ n ≤ 10⁵

---

#### 🚀 A tener en cuenta

* La solución más directa tiene complejidad **O(n log n)** (contar bits para cada número).
* ¿Puedes hacerlo en **O(n)** y en **una sola pasada**?
* ¿Puedes hacerlo **sin usar funciones integradas** como `bin()` o `__builtin_popcount()`?

# Ejercicio 2


### 🧩 **Enunciado**

Dada una **matriz `grid` de tamaño m x n** llena de **números no negativos**, encuentra un **camino desde la esquina superior izquierda hasta la esquina inferior derecha** tal que la **suma total de los números en el camino sea mínima**.

Solo puedes moverte **hacia la derecha** o **hacia abajo** en cada paso.

---

### ⚙️ **Restricciones**

* `m == grid.length`
* `n == grid[i].length`
* `1 <= m, n <= 200`
* `0 <= grid[i][j] <= 200`

---

### 💡 **Entrada**

Una matriz bidimensional `grid` que representa el costo de cada celda.

### 📤 **Salida**

Un entero que representa la **suma mínima posible** para llegar desde la esquina superior izquierda hasta la inferior derecha.

---

### 🧠 **Ejemplo 1**

**Entrada:**

```python
grid = [
  [1,3,1],
  [1,5,1],
  [4,2,1]
]
```

**Salida:**

```python
7
```

**Explicación:**
El camino con suma mínima es `1 → 3 → 1 → 1 → 1`, cuya suma es **7**.

---

### 🧠 **Ejemplo 2**

**Entrada:**

```python
grid = [
  [1,2,3],
  [4,5,6]
]
```

**Salida:**

```python
12
```

**Explicación:**
El camino mínimo corresponde a `1 → 2 → 3 → 6`, con una suma total de **12**.



# Ejercicio 3

📘 **1641. Conteo de Cadenas de Vocales Ordenadas (Count Sorted Vowel Strings)**
---

### 🧩 **Enunciado**

Dado un **entero `n`**, devuelve el **número de cadenas posibles de longitud `n`** que están **formadas únicamente por vocales** (`a`, `e`, `i`, `o`, `u`) y que además están **ordenadas lexicográficamente**.

Una cadena es **lexicográficamente ordenada** si, para todo índice válido `i`, se cumple que `s[i]` es **igual o anterior alfabéticamente** que `s[i+1]`.
Por ejemplo:
✅ `"ae"` es válida porque `'a' ≤ 'e'`
❌ `"ea"` no es válida porque `'e' > 'a'`

---

### ⚙️ **Restricciones**

* `1 <= n <= 50`

---

### 💡 **Entrada**

Un entero `n` que representa la **longitud de las cadenas** que deben construirse.

### 📤 **Salida**

Un entero que representa el **número total de cadenas posibles** de longitud `n` que cumplen las condiciones.

---

### 🧠 **Ejemplo 1**

**Entrada:**

```python
n = 1
```

**Salida:**

```python
5
```

**Explicación:**
Las 5 cadenas posibles son:
`["a", "e", "i", "o", "u"]`

---

### 🧠 **Ejemplo 2**

**Entrada:**

```python
n = 2
```

**Salida:**

```python
15
```

**Explicación:**
Las 15 cadenas posibles son:
`["aa", "ae", "ai", "ao", "au", "ee", "ei", "eo", "eu", "ii", "io", "iu", "oo", "ou", "uu"]`
Nota: `"ea"` no es válida porque `'e'` viene después de `'a'`.

---

### 🧠 **Ejemplo 3**

**Entrada:**

```python
n = 33
```

**Salida:**

```python
66045
```
