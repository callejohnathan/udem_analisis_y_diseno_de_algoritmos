## 🟥 140 · Word Break II — [1.7:0.5:-0.2]

**Enunciado**

Dada una cadena `s` y un diccionario de palabras `wordDict`, inserta espacios en `s` para construir una oración donde cada palabra sea una palabra válida del diccionario. Retorna todas las oraciones posibles en cualquier orden. Una misma palabra del diccionario puede reutilizarse tantas veces como sea necesario.

**Firma**
```python
def wordBreak(s: str, wordDict: List[str]) -> List[str]
```

**Casos de prueba**

| # | `s` | `wordDict` | Salida esperada |
|---|-----|------------|-----------------|
| 1 | `"catsanddog"` | `["cat","cats","and","sand","dog"]` | `["cat sand dog", "cats and dog"]` |
| 2 | `"catsandog"` | `["cats","dog","sand","and","cat"]` | `[]` *(no hay segmentación válida)* |

## 🟨 93 · Restore IP Addresses — [1:0.5:-0.1]

**Enunciado**

Dada una cadena `s` que contiene únicamente dígitos, retorna todas las direcciones IP válidas posibles que se pueden formar insertando puntos en `s`. No se permite reordenar ni eliminar ningún dígito de `s`. Una dirección IP válida consiste en exactamente cuatro enteros separados por puntos. Cada entero debe estar entre 0 y 255 (inclusive) y no puede tener ceros a la izquierda.

**Firma**
```python
def restoreIpAddresses(s: str) -> List[str]
```

**Casos de prueba**

| # | `s` | Salida esperada |
|---|-----|-----------------|
| 1 | `"25525511135"` | `["255.255.11.135", "255.255.111.35"]` |
| 2 | `"0000"` | `["0.0.0.0"]` *(solo los ceros son válidos, "00" no lo es)* |

## 🟨 491 · Non-decreasing Subsequences — [0.5:0:-0.1]

**Enunciado**

Dado un arreglo de enteros `nums`, retorna todas las subsecuencias no decrecientes distintas del arreglo que tengan al menos dos elementos. Puedes retornar la respuesta en cualquier orden. Las subsecuencias se forman respetando el orden original de los índices (no se reordena el arreglo).

**Firma**
```python
def findSubsequences(nums: List[int]) -> List[List[int]]
```

**Casos de prueba**

| # | `nums` | Salida esperada |
|---|--------|-----------------|
| 1 | `[4, 6, 7, 7]` | `[[4,6],[4,6,7],[4,6,7,7],[4,7],[4,7,7],[6,7],[6,7,7],[7,7]]` |
| 2 | `[4, 4, 3, 2, 1]` | `[[4,4]]` *(el único par no decreciente posible)* |
