# 🧩 Punto: Solucionador de Sudoku - [1.7:0.5:-0.2]

## 📌 Enunciado

Implemente un algoritmo que permita **resolver automáticamente un tablero de Sudoku** completando todas las casillas vacías.

El tablero está representado como una matriz de `9 x 9`, donde:

* Cada posición contiene un carácter entre `'1'` y `'9'`, o
* El carácter `'.'`, que representa una celda vacía.

Su programa debe modificar el tablero original de forma que quede completamente resuelto, cumpliendo todas las reglas del Sudoku.

---

## ✅ Reglas del Sudoku

Una solución válida debe satisfacer simultáneamente las siguientes condiciones:

1. **Cada fila** debe contener los dígitos del `1` al `9` **exactamente una vez**.
2. **Cada columna** debe contener los dígitos del `1` al `9` **exactamente una vez**.
3. Cada una de las **9 subcuadrículas de tamaño `3 x 3`** debe contener los dígitos del `1` al `9` **exactamente una vez**.

---

## 🎯 Objetivo

Dado un tablero parcialmente lleno, complete todas las posiciones vacías de forma que el resultado final sea un **Sudoku válido y completamente resuelto**.

---

## 📥 Entrada

Una matriz `board` de tamaño `9 x 9` de tipo `char`, donde:

* `'1'` a `'9'` representan valores ya ubicados.
* `'.'` representa una celda vacía.

---

## 📤 Salida

El tablero `board` debe quedar **modificado en sitio (in-place)**, conteniendo la solución completa del Sudoku.

---

## ⚠️ Restricciones

* El tablero siempre tendrá tamaño `9 x 9`.
* Se garantiza que existe **una única solución válida** para cada caso de prueba.
* No debe crearse una nueva solución como salida separada: la solución debe escribirse directamente sobre el tablero recibido.

# 🧩 Punto: Suma de combinaciones - [0.4:0.2:-0.1]

## 📌 Enunciado

Dado un arreglo de enteros **distintos** llamado `candidates` y un entero positivo `target`, implemente un algoritmo que retorne **todas las combinaciones únicas posibles** de elementos de `candidates` cuya suma sea exactamente igual a `target`.

Cada número del arreglo puede ser utilizado **un número ilimitado de veces** dentro de una misma combinación.

Dos combinaciones se consideran diferentes si al menos una tiene una **frecuencia distinta** de algún número respecto a la otra.

---

## 🎯 Objetivo

Encontrar todas las combinaciones válidas de números que sumen exactamente `target`, sin repetir soluciones equivalentes.

---

## 📥 Entrada

* Un arreglo de enteros distintos `candidates`.
* Un entero `target`.

---

## 📤 Salida

Una lista de listas, donde cada sublista representa una combinación válida de números cuya suma es exactamente `target`.

> El orden de las combinaciones en la salida **no es relevante**.

---

## ⚠️ Restricciones

* Todos los valores de `candidates` son **distintos**.
* Todos los valores son enteros positivos.
* Un mismo número puede ser usado múltiples veces.
* No deben aparecer combinaciones duplicadas.
* Se garantiza que la cantidad de soluciones válidas es menor que 150.

---

# 🧪 Ejemplos

## Ejemplo 1

### Entrada

```python
candidates = [2, 3, 6, 7]
target = 7
```

### Salida esperada

```python
[[2, 2, 3], [7]]
```

### Explicación

* `2 + 2 + 3 = 7`
* `7 = 7`

Estas son las únicas combinaciones posibles.

---

## Ejemplo 2

### Entrada

```python
candidates = [2, 3, 5]
target = 8
```

### Salida esperada

```python
[[2, 2, 2, 2], [2, 3, 3], [3, 5]]
```

### Explicación

Las combinaciones válidas que suman 8 son:

* `2 + 2 + 2 + 2`
* `2 + 3 + 3`
* `3 + 5`

---

## Ejemplo 3

### Entrada

```python
candidates = [2]
target = 1
```

### Salida esperada

```python
[]
```

### Explicación

No existe ninguna combinación posible que sume exactamente `1`.

---

# 🧠 Consideraciones importantes

* Si una combinación ya usa cierto número, puede volver a usarlo.
* No se deben considerar como diferentes combinaciones que solo cambian el orden.

### Ejemplo:

```python
[2, 2, 3]
```

y

```python
[3, 2, 2]
```

representan **la misma combinación**, por lo tanto solo debe aparecer una vez.
