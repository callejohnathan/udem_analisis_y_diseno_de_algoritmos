## Punto 1: Subconjuntos de Tamaño 5 sin "1"s Consecutivos

Dado un string de 0's y 1's de tamaño $N > 4$, encuentre cuántos subconjuntos de **exactamente 5 caracteres** se pueden formar tales que, al mantener su orden relativo original, **no contengan dos '1' seguidos** (el patrón "11").

> **Nota:** Un subconjunto se define por la combinación de los índices elegidos. Si hay caracteres idénticos en diferentes posiciones, cada combinación de índices cuenta como un subconjunto distinto.

* **Ejemplo 1 (N=6):**
* **Entrada:** `"110101"`
* **Subconjuntos posibles de tamaño 5:**
1. Índices {0,1,2,3,4} $\rightarrow$ `"11010"` (Contiene "11", **Inválido**)
2. Índices {0,1,2,3,5} $\rightarrow$ `"11011"` (Contiene "11", **Inválido**)
3. Índices {0,1,2,4,5} $\rightarrow$ `"11001"` (Contiene "11", **Inválido**)
4. Índices {0,2,3,4,5} $\rightarrow$ `"10101"` (**Válido**)
5. Índices {1,2,3,4,5} $\rightarrow$ `"10101"` (**Válido**)
* **Salida:** `2`

## ✅ Solución (Fuerza Bruta Iterativa)
```python
from typing import List


def count_valid_subsets_len5_no_11(s: str) -> int:
    """
    Cuenta cuántos subconjuntos (combinaciones de índices) de tamaño EXACTO 5
    se pueden formar de un string binario, manteniendo orden relativo,
    tal que el subsequence NO contenga el patrón "11".
    """
    n: int = len(s)                           # O(1) tiempo, O(1) espacio

    if n < 5:                                # O(1) tiempo, O(1) espacio
        return 0                             # O(1) tiempo, O(1) espacio

    count: int = 0                           # O(1) tiempo, O(1) espacio

    # Elegimos 5 índices i<j<k<l<m (combinación de 5 elementos)
    for i in range(0, n - 4):                # ∑ iteraciones = C(n,1) acotado; en total generará C(n,5)
        # por iteración: O(1) espacio extra
        for j in range(i + 1, n - 3):        # genera combinaciones ordenadas; contribuye a C(n,5)
            for k in range(j + 1, n - 2):    # contribuye a C(n,5)
                for l in range(k + 1, n - 1):# contribuye a C(n,5)
                    for m in range(l + 1, n):# contribuye a C(n,5)

                        # Construimos el subsequence de longitud 5
                        # Accesos a string por índice: O(1) cada uno
                        a: str = s[i]        # O(1) tiempo, O(1) espacio
                        b: str = s[j]        # O(1) tiempo, O(1) espacio
                        c: str = s[k]        # O(1) tiempo, O(1) espacio
                        d: str = s[l]        # O(1) tiempo, O(1) espacio
                        e: str = s[m]        # O(1) tiempo, O(1) espacio

                        # Verificamos que NO haya dos '1' seguidos
                        # Son 4 comparaciones constantes (longitud fija = 5)
                        is_valid: bool = True                 # O(1) tiempo, O(1) espacio

                        if a == "1" and b == "1":            # O(1) tiempo, O(1) espacio
                            is_valid = False                 # O(1) tiempo, O(1) espacio
                        if b == "1" and c == "1":            # O(1) tiempo, O(1) espacio
                            is_valid = False                 # O(1) tiempo, O(1) espacio
                        if c == "1" and d == "1":            # O(1) tiempo, O(1) espacio
                            is_valid = False                 # O(1) tiempo, O(1) espacio
                        if d == "1" and e == "1":            # O(1) tiempo, O(1) espacio
                            is_valid = False                 # O(1) tiempo, O(1) espacio

                        if is_valid:                          # O(1) tiempo, O(1) espacio
                            count += 1                        # O(1) tiempo, O(1) espacio

    return count                                               # O(1) tiempo, O(1) espacio
```

* El algoritmo recorre **todas las combinaciones de 5 índices**:

  \binom{n}{5} = \frac{n(n-1)(n-2)(n-3)(n-4)}{120}
  ]
$

$
✅ **Tiempo total:** **Θ(C(n,5)) = Θ(n⁵)**

* Solo variables escalares (`count`, `a..e`, `is_valid`).
* No se almacenan combinaciones.

✅ **Espacio total:** **Θ(1)**

## Punto 2: Triple Mayor Diferencia en Permutaciones

Dado un arreglo de $N$ números enteros, genere todas las permutaciones posibles. Para cada permutación, calcule la **resta sucesiva** de sus elementos (empezando desde el primero de izquierda a derecha). Encuentre los tres valores resultantes más altos.

* **Operación:** Si la permutación es $[n_1, n_2, n_3, ..., n_k]$, el cálculo es $n_1 - n_2 - n_3 - ... - n_k$.
* **Entrada:** Un arreglo de enteros.
* **Salida:** Los tres resultados más altos en orden descendente.

**Ejemplo:**

* **Entrada:** `[5, 10, 2]`
* **Cálculos:**
* `[10, 5, 2]` $\rightarrow 10 - 5 - 2 = \mathbf{3}$
* `[10, 2, 5]` $\rightarrow 10 - 2 - 5 = \mathbf{3}$
* `[5, 2, 10]` $\rightarrow 5 - 2 - 10 = \mathbf{-7}$
* `[5, 10, 2]` $\rightarrow 5 - 10 - 2 = \mathbf{-7}$
* `[2, 5, 10]` $\rightarrow 2 - 5 - 10 = \mathbf{-13}$
* `[2, 10, 5]` $\rightarrow 2 - 10 - 5 = \mathbf{-13}$


* **Salida:** `3, 3, -7`

---

* **Ejemplo**  (Con números negativos):

* **Entrada:** `[3, -5, 1]`
* **Análisis de permutaciones (Resta sucesiva):**
1. `[3, -5, 1]` $\rightarrow 3 - (-5) - 1 = 3 + 5 - 1 = \mathbf{7}$
2. `[3, 1, -5]` $\rightarrow 3 - 1 - (-5) = 3 - 1 + 5 = \mathbf{7}$
3. `[1, 3, -5]` $\rightarrow 1 - 3 - (-5) = 1 - 3 + 5 = \mathbf{3}$
4. `[1, -5, 3]` $\rightarrow 1 - (-5) - 3 = 1 + 5 - 3 = \mathbf{3}$
5. `[-5, 3, 1]` $\rightarrow -5 - 3 - 1 = \mathbf{-9}$
6. `[-5, 1, 3]` $\rightarrow -5 - 1 - 3 = \mathbf{-9}$


* **Salida:** `7, 7, 3`

---


## ✅ Solución (Fuerza Bruta Iterativa con Swaping)
> Idea: usar el algoritmo iterativo de **siguiente permutación (lexicográfica)**:

1. Ordenamos (para empezar desde la menor y recorrer todas)
2. Repetimos: evaluar permutación → generar siguiente → hasta que no haya más.

```python
from typing import List


def _reverse_suffix(a: List[int], left: int, right: int) -> None:
    """
    Revierte a[left:right+1] in-place.
    """
    while left < right:                 # O(t) donde t = (right-left+1), espacio O(1)
        a[left], a[right] = a[right], a[left]  # O(1) tiempo, O(1) espacio
        left += 1                       # O(1) tiempo, O(1) espacio
        right -= 1                      # O(1) tiempo, O(1) espacio


def _next_permutation(a: List[int]) -> bool:
    """
    Transforma a en su siguiente permutación lexicográfica.
    Retorna False si ya estaba en la última permutación.
    """
    n: int = len(a)                     # O(1) tiempo, O(1) espacio
    if n <= 1:                          # O(1) tiempo, O(1) espacio
        return False                    # O(1) tiempo, O(1) espacio

    # 1) Encontrar el pivote i tal que a[i] < a[i+1] desde la derecha
    i: int = n - 2                      # O(1) tiempo, O(1) espacio
    while i >= 0 and a[i] >= a[i + 1]:  # O(n) en el peor caso, espacio O(1)
        i -= 1                          # O(1) tiempo, O(1) espacio

    if i < 0:                           # O(1) tiempo, O(1) espacio
        return False                    # O(1) tiempo, O(1) espacio

    # 2) Encontrar el menor elemento a[j] > a[i] a la derecha
    j: int = n - 1                      # O(1) tiempo, O(1) espacio
    while a[j] <= a[i]:                 # O(n) en el peor caso, espacio O(1)
        j -= 1                          # O(1) tiempo, O(1) espacio

    # 3) Swap pivote con a[j]
    a[i], a[j] = a[j], a[i]             # O(1) tiempo, O(1) espacio

    # 4) Revertir el sufijo i+1..n-1 para dejarlo mínimo
    _reverse_suffix(a, i + 1, n - 1)    # O(n) peor caso, espacio O(1)

    return True                          # O(1) tiempo, O(1) espacio


def top3_successive_subtraction_results(nums: List[int]) -> List[int]:
    """
    Genera todas las permutaciones de nums (fuerza bruta iterativa).
    Para cada permutación p = [n1,n2,...,nk], calcula n1-n2-...-nk.
    Retorna los 3 resultados más altos en orden descendente.
    """
    n: int = len(nums)                   # O(1) tiempo, O(1) espacio
    if n == 0:                           # O(1) tiempo, O(1) espacio
        return []                        # O(1) tiempo, O(1) espacio

    a: List[int] = nums[:]               # O(n) tiempo, O(n) espacio (copia para no mutar entrada)
    a.sort()                             # O(n log n) tiempo, O(1) extra (in-place)

    # Mantendremos top3 manualmente (sin guardar todos los resultados)
    # top[0] >= top[1] >= top[2]
    top: List[int] = []                  # O(1) tiempo, O(1) espacio (máximo 3 elementos)

    while True:                          # Se repite P veces, P = n! / (repeticiones por duplicados)
        # Calcular n1 - n2 - ... - nk
        # Para k=n, esto es O(n) por permutación.
        result: int = a[0]               # O(1) tiempo, O(1) espacio
        idx: int = 1                     # O(1) tiempo, O(1) espacio
        while idx < n:                   # O(n) por permutación, O(1) espacio
            result -= a[idx]             # O(1) tiempo, O(1) espacio
            idx += 1                     # O(1) tiempo, O(1) espacio

        # Insertar result en top manteniendo orden (lista tamaño <= 3)
        # Esto es O(1) porque el tamaño es constante (máximo 3).
        if len(top) == 0:                # O(1) tiempo, O(1) espacio
            top.append(result)           # O(1) tiempo, O(1) espacio
        elif len(top) == 1:              # O(1) tiempo, O(1) espacio
            if result >= top[0]:         # O(1) tiempo, O(1) espacio
                top.insert(0, result)    # O(1) tiempo (tamaño <= 2), O(1) espacio
            else:
                top.append(result)       # O(1) tiempo, O(1) espacio
        elif len(top) == 2:              # O(1) tiempo, O(1) espacio
            if result >= top[0]:
                top.insert(0, result)    # O(1) tiempo, O(1) espacio
            elif result >= top[1]:
                top.insert(1, result)    # O(1) tiempo, O(1) espacio
            else:
                top.append(result)       # O(1) tiempo, O(1) espacio
        else:  # len(top) == 3
            if result >= top[0]:
                top.insert(0, result)    # O(1) tiempo, O(1) espacio
                top.pop()                # O(1) tiempo, O(1) espacio
            elif result >= top[1]:
                top.insert(1, result)    # O(1) tiempo, O(1) espacio
                top.pop()                # O(1) tiempo, O(1) espacio
            elif result >= top[2]:
                top[2] = result          # O(1) tiempo, O(1) espacio

        # Ir a la siguiente permutación; si no hay, terminamos
        if not _next_permutation(a):     # O(n) peor caso por permutación, O(1) espacio
            break                        # O(1) tiempo, O(1) espacio

    # Si n! < 3 (ej: n=1 o n=2), devolvemos lo que haya
    return top                            # O(1) tiempo, O(1) espacio
```

## 📌 Complejidad total

Sea **P** el número de permutaciones generadas:

* Si todos los elementos son distintos: **P = n!**
* Si hay duplicados: **P = n! / (c1! c2! …)** (menos permutaciones únicas)

### ⏱️ Tiempo

Por cada permutación:

* calcular resta sucesiva: **O(n)**
* generar siguiente permutación: **O(n)** (peor caso)
* actualizar top3: **O(1)** (constante)

Entonces:
[
T(n)=\Theta(P \cdot n)
]
✅ Si todos distintos: **Θ(n · n!)**

### 💾 Espacio

* Copia del arreglo: **O(n)**
* Variables auxiliares y top3: **O(1)**

✅ **Espacio total:** **Θ(n)**

````python
from typing import List


def _reverse_suffix(a: List[int], left: int, right: int) -> None:
    """
    Revierte a[left:right+1] in-place.
    """
    while left < right:                 # O(t) donde t = (right-left+1), espacio O(1)
        a[left], a[right] = a[right], a[left]  # O(1) tiempo, O(1) espacio
        left += 1                       # O(1) tiempo, O(1) espacio
        right -= 1                      # O(1) tiempo, O(1) espacio


def _next_permutation(a: List[int]) -> bool:
    """
    Transforma a en su siguiente permutación lexicográfica.
    Retorna False si ya estaba en la última permutación.
    """
    n: int = len(a)                     # O(1) tiempo, O(1) espacio
    if n <= 1:                          # O(1) tiempo, O(1) espacio
        return False                    # O(1) tiempo, O(1) espacio

    # 1) Encontrar el pivote i tal que a[i] < a[i+1] desde la derecha
    i: int = n - 2                      # O(1) tiempo, O(1) espacio
    while i >= 0 and a[i] >= a[i + 1]:  # O(n) en el peor caso, espacio O(1)
        i -= 1                          # O(1) tiempo, O(1) espacio

    if i < 0:                           # O(1) tiempo, O(1) espacio
        return False                    # O(1) tiempo, O(1) espacio

    # 2) Encontrar el menor elemento a[j] > a[i] a la derecha
    j: int = n - 1                      # O(1) tiempo, O(1) espacio
    while a[j] <= a[i]:                 # O(n) en el peor caso, espacio O(1)
        j -= 1                          # O(1) tiempo, O(1) espacio

    # 3) Swap pivote con a[j]
    a[i], a[j] = a[j], a[i]             # O(1) tiempo, O(1) espacio

    # 4) Revertir el sufijo i+1..n-1 para dejarlo mínimo
    _reverse_suffix(a, i + 1, n - 1)    # O(n) peor caso, espacio O(1)

    return True                          # O(1) tiempo, O(1) espacio


def top3_successive_subtraction_results(nums: List[int]) -> List[int]:
    """
    Genera todas las permutaciones de nums (fuerza bruta iterativa).
    Para cada permutación p = [n1,n2,...,nk], calcula n1-n2-...-nk.
    Retorna los 3 resultados más altos en orden descendente.
    """
    n: int = len(nums)                   # O(1) tiempo, O(1) espacio
    if n == 0:                           # O(1) tiempo, O(1) espacio
        return []                        # O(1) tiempo, O(1) espacio

    a: List[int] = nums[:]               # O(n) tiempo, O(n) espacio (copia para no mutar entrada)
    a.sort()                             # O(n log n) tiempo, O(1) extra (in-place)

    # Mantendremos top3 manualmente (sin guardar todos los resultados)
    # top[0] >= top[1] >= top[2]
    top: List[int] = []                  # O(1) tiempo, O(1) espacio (máximo 3 elementos)

    while True:                          # Se repite P veces, P = n! / (repeticiones por duplicados)
        # Calcular n1 - n2 - ... - nk
        # Para k=n, esto es O(n) por permutación.
        result: int = a[0]               # O(1) tiempo, O(1) espacio
        idx: int = 1                     # O(1) tiempo, O(1) espacio
        while idx < n:                   # O(n) por permutación, O(1) espacio
            result -= a[idx]             # O(1) tiempo, O(1) espacio
            idx += 1                     # O(1) tiempo, O(1) espacio

        # Insertar result en top manteniendo orden (lista tamaño <= 3)
        # Esto es O(1) porque el tamaño es constante (máximo 3).
        if len(top) == 0:                # O(1) tiempo, O(1) espacio
            top.append(result)           # O(1) tiempo, O(1) espacio
        elif len(top) == 1:              # O(1) tiempo, O(1) espacio
            if result >= top[0]:         # O(1) tiempo, O(1) espacio
                top.insert(0, result)    # O(1) tiempo (tamaño <= 2), O(1) espacio
            else:
                top.append(result)       # O(1) tiempo, O(1) espacio
        elif len(top) == 2:              # O(1) tiempo, O(1) espacio
            if result >= top[0]:
                top.insert(0, result)    # O(1) tiempo, O(1) espacio
            elif result >= top[1]:
                top.insert(1, result)    # O(1) tiempo, O(1) espacio
            else:
                top.append(result)       # O(1) tiempo, O(1) espacio
        else:  # len(top) == 3
            if result >= top[0]:
                top.insert(0, result)    # O(1) tiempo, O(1) espacio
                top.pop()                # O(1) tiempo, O(1) espacio
            elif result >= top[1]:
                top.insert(1, result)    # O(1) tiempo, O(1) espacio
                top.pop()                # O(1) tiempo, O(1) espacio
            elif result >= top[2]:
                top[2] = result          # O(1) tiempo, O(1) espacio

        # Ir a la siguiente permutación; si no hay, terminamos
        if not _next_permutation(a):     # O(n) peor caso por permutación, O(1) espacio
            break                        # O(1) tiempo, O(1) espacio

    # Si n! < 3 (ej: n=1 o n=2), devolvemos lo que haya
    return top                            # O(1) tiempo, O(1) espacio

print(top3_successive_subtraction_results([5, 10, 2]))   # [3, 3, -7]
print(top3_successive_subtraction_results([3, -5, 1]))   # [7, 7, 3]
````

## ✅ Solución: Permutaciones desde cero (iterativa, sin swap) + `typehints`

> 💡 Idea: simular backtracking pero **sin recursión** usando un “stack” de estados:

* `perm`: la permutación parcial
* `used`: cuáles índices ya se usaron
* `sum_so_far`: resultado parcial de la resta sucesiva

```python
from typing import List, Tuple


def top3_successive_subtraction_results_building(nums: List[int]) -> List[int]:
    """
    Genera todas las permutaciones de nums (fuerza bruta) construyéndolas desde cero,
    de manera ITERATIVA y SIN SWAP.
    Para cada permutación p=[n1,...,nk] calcula n1-n2-...-nk y retorna los 3 valores más altos.
    """
    n: int = len(nums)                                  # O(1) tiempo, O(1) espacio
    if n == 0:                                          # O(1) tiempo, O(1) espacio
        return []                                       # O(1) tiempo, O(1) espacio

    # top siempre será una lista ordenada desc con tamaño <= 3
    top: List[int] = []                                 # O(1) tiempo, O(1) espacio

    # Estado en stack:
    # (depth, used, current_value)
    # - depth: cuántos elementos lleva la permutación parcial
    # - used: vector booleano de tamaño n
    # - current_value: valor parcial de n1-n2-... hasta depth
    #
    # Nota: no guardamos la permutación completa porque para el resultado solo basta el valor.
    stack: List[Tuple[int, List[bool], int]] = []        # O(1) tiempo, O(1) espacio inicial

    # Inicialización: elegimos el primer elemento de la permutación (n opciones)
    for i in range(n):                                   # O(n) tiempo, O(1) extra por iteración
        used: List[bool] = [False] * n                   # O(n) tiempo, O(n) espacio
        used[i] = True                                   # O(1) tiempo, O(1) espacio
        stack.append((1, used, nums[i]))                 # O(1) tiempo, O(1) espacio (referencia)

    # Recorremos todos los estados (equivalente a explorar n! hojas)
    while len(stack) > 0:                                # se ejecuta O(#estados) veces
        depth, used, current_value = stack.pop()         # O(1) tiempo, O(1) espacio

        if depth == n:                                   # O(1) tiempo, O(1) espacio
            # Permutación completa: current_value es el resultado final
            result: int = current_value                  # O(1) tiempo, O(1) espacio

            # Insertar result en top (tamaño <= 3 => O(1))
            if len(top) == 0:                            # O(1)
                top.append(result)                       # O(1)
            elif len(top) == 1:                          # O(1)
                if result >= top[0]:                     # O(1)
                    top.insert(0, result)                # O(1) (tamaño <= 2)
                else:
                    top.append(result)                   # O(1)
            elif len(top) == 2:                          # O(1)
                if result >= top[0]:
                    top.insert(0, result)                # O(1)
                elif result >= top[1]:
                    top.insert(1, result)                # O(1)
                else:
                    top.append(result)                   # O(1)
            else:  # len(top) == 3
                if result >= top[0]:
                    top.insert(0, result)                # O(1)
                    top.pop()                            # O(1)
                elif result >= top[1]:
                    top.insert(1, result)                # O(1)
                    top.pop()                            # O(1)
                elif result >= top[2]:
                    top[2] = result                      # O(1)

            continue                                     # O(1) tiempo, O(1) espacio

        # Expandir: elegir el siguiente elemento no usado
        # Esto genera todas las permutaciones posibles (fuerza bruta).
        for i in range(n):                               # O(n) por estado
            if not used[i]:                              # O(1)
                # Creamos un nuevo estado (sin swap): copiamos used y marcamos i
                new_used: List[bool] = used[:]           # O(n) tiempo, O(n) espacio
                new_used[i] = True                       # O(1) tiempo, O(1) espacio

                # Actualizamos el valor parcial:
                # si depth>=1, siempre restamos el nuevo elemento (porque es n_{depth+1})
                new_value: int = current_value - nums[i] # O(1) tiempo, O(1) espacio

                stack.append((depth + 1, new_used, new_value))  # O(1) tiempo, O(1) espacio

    return top                                           # O(1) tiempo, O(1) espacio
```

---

## 🧠 Complejidad total (muy clara)

### ⏱️ Tiempo

* Se generan **todas** las permutaciones: **n!**
* En cada nivel/estado se itera sobre `n` y además se copia `used` de tamaño `n`.

La copia `used[:]` cuesta **O(n)**, y se hace por cada expansión.

En términos simples y correctos para parcial:

✅ **Tiempo total:**
$
[
T(n)=\Theta(n! \cdot n^2)
]
$
* `n!` por las permutaciones
* `n` por explorar candidatos en cada expansión
* `n` por copiar `used`

> Nota didáctica: si evitáramos copiar `used` (mutando y desmutando) bajaríamos un factor, pero eso ya sería “otra técnica”.

### 💾 Espacio

* `used` por estado ocupa O(n).
* El stack puede crecer mucho (dependiendo del orden de exploración). En el peor caso, puede acumular un número grande de estados.

✅ **Espacio (peor caso):** **O(n · #estados)** ≈ **O(n · n!)** (muy conservador).
✅ **Espacio mínimo de trabajo por ruta:** O(n) (profundidad), pero aquí usamos stack con copias.

````python
from typing import List, Tuple


def top3_successive_subtraction_results_building(nums: List[int]) -> List[int]:
    """
    Genera todas las permutaciones de nums (fuerza bruta) construyéndolas desde cero,
    de manera ITERATIVA y SIN SWAP.
    Para cada permutación p=[n1,...,nk] calcula n1-n2-...-nk y retorna los 3 valores más altos.
    """
    n: int = len(nums)                                  # O(1) tiempo, O(1) espacio
    if n == 0:                                          # O(1) tiempo, O(1) espacio
        return []                                       # O(1) tiempo, O(1) espacio

    # top siempre será una lista ordenada desc con tamaño <= 3
    top: List[int] = []                                 # O(1) tiempo, O(1) espacio

    # Estado en stack:
    # (depth, used, current_value)
    # - depth: cuántos elementos lleva la permutación parcial
    # - used: vector booleano de tamaño n
    # - current_value: valor parcial de n1-n2-... hasta depth
    #
    # Nota: no guardamos la permutación completa porque para el resultado solo basta el valor.
    stack: List[Tuple[int, List[bool], int]] = []        # O(1) tiempo, O(1) espacio inicial

    # Inicialización: elegimos el primer elemento de la permutación (n opciones)
    for i in range(n):                                   # O(n) tiempo, O(1) extra por iteración
        used: List[bool] = [False] * n                   # O(n) tiempo, O(n) espacio
        used[i] = True                                   # O(1) tiempo, O(1) espacio
        stack.append((1, used, nums[i]))                 # O(1) tiempo, O(1) espacio (referencia)

    # Recorremos todos los estados (equivalente a explorar n! hojas)
    while len(stack) > 0:                                # se ejecuta O(#estados) veces
        depth, used, current_value = stack.pop()         # O(1) tiempo, O(1) espacio

        if depth == n:                                   # O(1) tiempo, O(1) espacio
            # Permutación completa: current_value es el resultado final
            result: int = current_value                  # O(1) tiempo, O(1) espacio

            # Insertar result en top (tamaño <= 3 => O(1))
            if len(top) == 0:                            # O(1)
                top.append(result)                       # O(1)
            elif len(top) == 1:                          # O(1)
                if result >= top[0]:                     # O(1)
                    top.insert(0, result)                # O(1) (tamaño <= 2)
                else:
                    top.append(result)                   # O(1)
            elif len(top) == 2:                          # O(1)
                if result >= top[0]:
                    top.insert(0, result)                # O(1)
                elif result >= top[1]:
                    top.insert(1, result)                # O(1)
                else:
                    top.append(result)                   # O(1)
            else:  # len(top) == 3
                if result >= top[0]:
                    top.insert(0, result)                # O(1)
                    top.pop()                            # O(1)
                elif result >= top[1]:
                    top.insert(1, result)                # O(1)
                    top.pop()                            # O(1)
                elif result >= top[2]:
                    top[2] = result                      # O(1)

            continue                                     # O(1) tiempo, O(1) espacio

        # Expandir: elegir el siguiente elemento no usado
        # Esto genera todas las permutaciones posibles (fuerza bruta).
        for i in range(n):                               # O(n) por estado
            if not used[i]:                              # O(1)
                # Creamos un nuevo estado (sin swap): copiamos used y marcamos i
                new_used: List[bool] = used[:]           # O(n) tiempo, O(n) espacio
                new_used[i] = True                       # O(1) tiempo, O(1) espacio

                # Actualizamos el valor parcial:
                # si depth>=1, siempre restamos el nuevo elemento (porque es n_{depth+1})
                new_value: int = current_value - nums[i] # O(1) tiempo, O(1) espacio

                stack.append((depth + 1, new_used, new_value))  # O(1) tiempo, O(1) espacio

    return top                                           # O(1) tiempo, O(1) espacio

print(top3_successive_subtraction_results_building([5, 10, 2]))   # [3, 3, -7]
print(top3_successive_subtraction_results_building([3, -5, 1]))   # [7, 7, 3]
````

## Solución 3 --> la más común entre ustedes

````python
def obtener_permutaciones(arr):
    """
    Función iterativa para calcular todas las permutaciones de una lista
    utilizando el método de inserción (sin usar intercambios/swapping).

    Complejidad Temporal Total: O(n * n!) - n es la longitud de arr.
    Complejidad Espacial Total: O(n * n!) - para almacenar todas las permutaciones.
    """
    # Caso base: O(1)
    if not arr:
        return [[]]

    # Estado inicial: O(1)
    permutaciones = [[]]

    # Bucle exterior: se ejecuta 'n' veces (O(n))
    for elemento_actual in arr:
        siguientes_permutaciones = []

        # Bucle intermedio: se ejecuta i! veces (donde i es el progreso actual)
        for permutacion_actual in permutaciones:

            # Bucle interior: se ejecuta i+1 veces (O(i))
            for posicion in range(len(permutacion_actual) + 1):

                # Slicing y concatenación: O(i) - Crea una nueva lista de tamaño i+1
                parte_izquierda = permutacion_actual[:posicion]
                parte_derecha = permutacion_actual[posicion:]
                nueva_permutacion = parte_izquierda + [elemento_actual] + parte_derecha

                # Agregar a la lista: O(1) amortizado
                siguientes_permutaciones.append(nueva_permutacion)

        # Reemplazo de lista: O(1)
        permutaciones = siguientes_permutaciones

    return permutaciones

def calcular_mayores_diferencias(permutaciones):
    """
    Calcula la resta sucesiva de los elementos de cada permutación
    y devuelve los 3 resultados más altos.

    Complejidad Temporal: O(n * n!) - Recorre n! listas de tamaño n.
    Complejidad Espacial: O(n!) - Almacena los resultados de cada resta.
    """
    resultados = []

    # Recorrer n! permutaciones: O(n!)
    for perm in permutaciones:
        if not perm:
            continue

        # Resta sucesiva: O(n) por cada permutación
        res = perm[0]
        for i in range(1, len(perm)):
            res -= perm[i]
        resultados.append(res)

    # Ordenamiento: O(N log N) donde N = n!. En términos de n: O(n! log(n!))
    resultados_ordenados = sorted(resultados, reverse=True)

    # Retorno del top 3: O(1)
    return resultados_ordenados[:3]

# ==========================================
# Análisis de Complejidad (Resumen)
# ==========================================
# 1. obtener_permutaciones:
#    - Temporal: O(n * n!). El número de permutaciones es n!, y construir cada
#      una mediante inserción/slicing toma tiempo proporcional a n.
#    - Espacial: O(n * n!). Se deben guardar n! listas, cada una con n elementos.
#
# 2. calcular_mayores_diferencias:
#    - Temporal: O(n * n!). Procesar cada lista toma O(n). El ordenamiento
#      final es O(n! log n!).
#    - Espacial: O(n!). Se guarda un escalar por cada permutación procesada.

if __name__ == "__main__":
    mi_array = [10, 5, 2]
    print(f"Calculando para el array: {mi_array}")

    lista_permutaciones = obtener_permutaciones(mi_array)
    print(f"Total de permutaciones: {len(lista_permutaciones)}")

    top_3_diferencias = calcular_mayores_diferencias(lista_permutaciones)

    print("\nLos 3 resultados mayores de la resta sucesiva (e1 - e2 - e3 - ...):")
    for i, valor in enumerate(top_3_diferencias):
        print(f"{i + 1}: {valor}")
````

## Punto 3: Subarreglos Continuos con Límite $X$

Dado un arreglo de enteros no negativos y un valor entero $X$, cuente cuántos **subarreglos continuos** tienen una suma que sea menor o igual a $X$.

* **Entrada:** Un arreglo `A` y un entero `X`.
* **Salida:** Cantidad de subarreglos que cumplen la condición.

**Ejemplo:**

* **Entrada:** `A = [1, 2, 3]`, `X = 3`
* **Salida:** `4`
*(Explicación: Los subarreglos continuos que cumplen son: `[1]`, `[2]`, `[3]` y `[1, 2]`. El subarreglo `[2, 3]` suma 5 y `[1, 2, 3]` suma 6, ambos superan $X=3$).*


---

## ✅ Solución (Fuerza Bruta Iterativa) con `typehints` + complejidad por línea

```python
from typing import List


def count_subarrays_sum_leq_x(a: List[int], x: int) -> int:
    """
    Cuenta cuántos subarreglos continuos tienen suma <= x.
    Fuerza bruta iterativa: explora todos los (i, j) con i <= j.
    """
    n: int = len(a)                           # O(1) tiempo, O(1) espacio
    count: int = 0                            # O(1) tiempo, O(1) espacio

    for i in range(n):                        # O(n) iteraciones, O(1) espacio extra
        current_sum: int = 0                  # O(1) tiempo, O(1) espacio
        for j in range(i, n):                 # O(n-i) iteraciones; total ~ O(n^2)
            current_sum += a[j]               # O(1) tiempo, O(1) espacio

            if current_sum <= x:              # O(1) tiempo, O(1) espacio
                count += 1                    # O(1) tiempo, O(1) espacio
            # Nota: No hacemos break aunque podríamos (por no-negatividad),
            # porque el enunciado pide enfoque fuerza bruta.

    return count                              # O(1) tiempo, O(1) espacio
```

---

## 📌 Complejidad total

* Hay $(\frac{n(n+1)}{2})$ subarreglos continuos.
* Con esta versión, cada subarreglo se “visita” una vez mediante acumulación incremental.

### ⏱️ Tiempo

[
$T(n) = \Theta(n^2)$
]

### 💾 Espacio

[
$S(n) = \Theta(1)$
]
