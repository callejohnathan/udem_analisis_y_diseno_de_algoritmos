Aquí tienes **3 ejercicios de fuerza bruta** (de introductorio → retador) y **3 ejercicios de conteo** (más difíciles) que **obligan** a usar **diccionarios/hashmaps**. Los enuncio en estilo competitivo (input/output) y con restricciones pensadas para que *primero* sea viable por brute force y luego se preste para discutir mejoras.

---

## ✅ Fuerza bruta (3)

### 1) Par con suma más cercana

**Nivel:** Intro
Dado un arreglo de enteros, encuentra el par de índices distintos ((i, j)) cuya suma (a_i + a_j) sea **más cercana** a un valor (T).
Si hay empate, elige el par con **menor suma absoluta** (|a_i + a_j|). Si persiste el empate, el menor (i), luego menor (j).

**Entrada**

* (n) (2 ≤ n ≤ 2000)
* (n) enteros
* (T)

**Salida**

* (i) (j) (0-index o 1-index, tú defines en clase), y la suma del par

**Brute force esperado:** doble loop (O(n^2))

---

### 2) Subarreglo con suma exacta y longitud mínima

**Nivel:** Medio
Dado un arreglo de enteros (pueden ser negativos), encuentra un subarreglo continuo cuya suma sea exactamente (S) y que tenga **longitud mínima**.
Si no existe, imprime `-1`.

**Entrada**

* (n) (1 ≤ n ≤ 2000)
* (n) enteros (pueden ser negativos)
* (S)

**Salida**

* (l) (r) (índices del subarreglo) o `-1`

**Brute force esperado:** probar todos los ((l, r)) con suma acumulada interna (O(n^2)) o (O(n^3)) si lo hacen “ingenuo” (y ahí haces la discusión).

---

### 3) Mejor “rectángulo” en una matriz binaria

**Nivel:** Retador (pero sigue brute force)
Dada una matriz (N \times M) de 0/1, encuentra el rectángulo (submatriz) con **máxima cantidad de 1s**.
Si hay empate, elige el de **menor área**; si persiste, el que tenga esquina superior izquierda “más pequeña” (fila, luego columna).

**Entrada**

* (N, M) (1 ≤ N, M ≤ 60)
* (N) líneas con (M) caracteres `0/1`

**Salida**

* Coordenadas del rectángulo: (r1, c1, r2, c2) y el número de 1s

**Brute force esperado:** enumerar todos los rectángulos (O(N^2 M^2)) y contar 1s con loops internos (te queda pesado si cuentan mal, perfecto para hablar de optimización más adelante con prefijos 2D).

## 🔥 Conteo con diccionarios (3) — mayor dificultad

### 4) Subarreglos con suma divisible por K

**Nivel:** Intermedio-alto
Dado un arreglo de enteros y un entero (K), cuenta cuántos subarreglos tienen suma divisible por (K).

**Entrada**

* (n) (1 ≤ n ≤ 2e5)
* (n) enteros (pueden ser negativos)
* (K) (2 ≤ K ≤ 2e5)

**Salida**

* Un entero: cantidad de subarreglos

**Idea obligatoria:** diccionario de frecuencias de residuos de prefijos (hashmap).
*(Aquí se ve brutalmente el poder del conteo por equivalencias.)*

---

### 5) Pares “buenos” por diferencia fija

**Nivel:** Alto
Dado un arreglo y un entero (D), cuenta la cantidad de pares ((i, j)) con (i < j) tales que (a_j - a_i = D).

**Entrada**

* (n) (1 ≤ n ≤ 2e5)
* (n) enteros (valor absoluto hasta 1e9)
* (D) (puede ser negativo,  |D| ≤ 1e9)

**Salida**

* Cantidad de pares

**Idea obligatoria:** diccionario de conteo visto-hasta-ahora.
(Se puede resolver en (O(n)), y es perfecto para discutir duplicados y frecuencias.)

---

### 6) Anagramas por ventana

**Nivel:** Muy alto (pero hermoso)
Dadas dos cadenas (S) y (P), cuenta cuántas posiciones (i) existen tales que el substring (S[i..i+|P|-1]) es un **anagrama** de (P).
Las cadenas pueden tener letras minúsculas y mayúsculas (o solo minúsculas si quieres simplificar).

**Entrada**

* Cadena (S) (1 ≤ |S| ≤ 2e5)
* Cadena (P) (1 ≤ |P| ≤ |S|)

**Salida**

* Un entero: número de ventanas que son anagrama

**Idea obligatoria:** diccionarios de frecuencia + actualización por ventana (sliding).
*(Esto es conteo con diccionario en modo serio.)*
