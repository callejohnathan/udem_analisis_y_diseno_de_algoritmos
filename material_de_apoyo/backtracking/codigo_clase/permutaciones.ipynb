```python
from typing import List
def permutations(string: str) -> List[str]:
    #asumiendo que el string tiene 3 caracteres
    n: int = len(string)
    permutations: List[str] = []
    for i in range(n):
        for j in range(n):
            if(j != i):
                for k in range(n):
                    if j != k and i != k:
                        permutations.append(string[i] + string[j] + string[k])
    
    return permutations

print(permutations("abc"))  # Output: ['abc', 'acb', 'bac', 'bca', 'cab', 'cba']
```

```python
def iterative_permutation(string: str)-> List[str]:
    n: int = len(string)
    permutations: List[str] = []
    queue: list[str] = list(string)
    while(queue):
        print(queue)
        current = queue.pop(0)
        if len(current) == n:
            permutations.append("".join(current))
        else:
            for i in range(len(queue)):
                new_permutation = current[:i] + current[i+1:]
                queue.append(new_permutation)
    
    return permutations

print(iterative_permutation("abc"))  # Output: ['abc', 'acb', 'bac', 'bca', 'cab', 'cba']
```
