## Punto 1 — Tramo de montaña más largo

- Tiene al menos 3 elementos.
- Existe una posición `i`, con `0 < i < longitud − 1`, tal que:
  - `arr[0] < arr[1] < ... < arr[i − 1] < arr[i]`
