Una aplicación de música desea generar automáticamente una lista de reproducción personalizada a partir del historial de escuchas de un usuario.
El historial está representado por un string `S` de longitud `n > 1000`, donde **cada carácter** identifica una canción reproducida (pueden repetirse canciones a lo largo del historial).

El sistema debe generar una lista de reproducción `M`, representada con un string, que contenga **todas las canciones distintas** del historial, ordenadas de acuerdo con una **heurística de relevancia** que combine dos factores:

1. **Frecuencia de escucha (`freq(c)`):** número de veces que la canción fue reproducida.
2. **Última aparición (`last(c)`):** posición en la que la canción fue escuchada por última vez.

---
