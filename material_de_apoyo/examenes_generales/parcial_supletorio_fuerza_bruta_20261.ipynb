## Punto 1 — Combinatoria
### Contexto

Una farmacia necesita generar todas las posibles combinaciones de medicamentos que se pueden incluir en un kit de primeros auxilios. Dado un catálogo de `n` medicamentos y un tamaño de kit `k`, el sistema debe listar todas las combinaciones posibles de exactamente `k` medicamentos sin importar el orden, y sin repetir el mismo medicamento dos veces en el mismo kit.

### Enunciado

Dada una lista de enteros `medicamentos` (sin duplicados) y un entero `k`, retorna todas las combinaciones posibles de exactamente `k` elementos. El orden dentro de cada combinación no importa, y cada elemento solo puede aparecer una vez por combinación. Puedes retornar las combinaciones en cualquier orden.


### Casos de prueba

| # | `medicamentos` | `k` | Salida esperada |
|---|----------------|-----|-----------------|
| 1 | `[1, 2, 3, 4]` | `2` | `[[1,2], [1,3], [1,4], [2,3], [2,4], [3,4]]` |
| 2 | `[10, 20, 30]` | `3` | `[[10, 20, 30]]` |

> **Nota:** los medicamentos están representados por sus códigos numéricos. El orden de la salida no se evalúa, pero cada combinación no debe repetirse.

## Punto 2 — Permutaciones

### Contexto

Un sistema de seguridad genera contraseñas temporales a partir de un conjunto fijo de caracteres. Por razones de auditoría, el equipo necesita saber cuántas de todas las posibles contraseñas generadas con esos caracteres resultan ser palíndromos, es decir, se leen igual de izquierda a derecha que de derecha a izquierda.

### Enunciado

Dado un string `s`, cuenta cuántas de sus permutaciones únicas son palindrómicas. Debes generar las permutaciones usando recursión y backtracking. No está permitido usar `itertools` ni funciones similares.


### Casos de prueba

| # | `s` | Salida esperada | Explicación |
|---|-----|-----------------|-------------|
| 1 | `"aab"` | `1` | Las permutaciones únicas son `"aab"`, `"aba"`, `"baa"`. Solo `"aba"` es palíndromo. |
| 2 | `"aaa"` | `1` | Solo existe una permutación única: `"aaa"`, que es palíndromo. |

> **Nota:** si el string tiene caracteres repetidos, permutaciones idénticas solo se cuentan una vez. Por ejemplo, para `"aa"` la única permutación es `"aa"` y se cuenta como `1`.

## Punto 3 — Exploración de matrices

### Contexto

Un sistema de análisis financiero recibe una cuadrícula de valores numéricos que representan transacciones registradas en distintas sucursales durante un mes. El equipo de auditoría necesita identificar todos los grupos de exactamente tres celdas cuya suma sea igual a un monto objetivo `K`, para investigar posibles irregularidades.

### Enunciado

Dada una matriz de enteros de `n × m` y un entero `K`, encuentra todas las tripletas de celdas cuya suma sea exactamente `K`. Las tres celdas deben ser distintas (no puedes usar la misma celda dos veces en una tripleta), pero pueden estar en cualquier posición de la matriz. Retorna una lista de tripletas, donde cada tripleta es una lista de tres tuplas `(fila, columna)`. Para evitar duplicados, reporta cada tripleta una sola vez: considera que `[A, B, C]` y `[B, A, C]` son la misma tripleta.


### Casos de prueba

**Caso 1**
```python
matriz = [
    [1, 2, 3],
    [4, 5, 6]
]
K = 6
```
Salida esperada:
```python
[
    [(0,0), (0,1), (0,2)] # 1 + 2 + 3 = 6
]
```
> Retorna todas las tripletas de celdas distintas que sumen exactamente `6`. Por ejemplo, `(0,0), (0,1), (0,2)` → `1 + 2 + 3 = 6` ✓

**Caso 2**
```python
matriz = [
    [1, 1],
    [1, 1]
]
K = 10
```
Salida esperada: `[]` *(ninguna tripleta suma 10)*

> **Nota:** el orden de las tripletas en la salida no se evalúa. No reportes la misma combinación de celdas más de una vez.
