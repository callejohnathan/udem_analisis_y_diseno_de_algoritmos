## 📝 Ejercicio 1

Un **punto fijo** en un arreglo $A[0..N]$ se define como un índice $i$ tal que:

$$
A[i] = i
$$

Por ejemplo:

* En el arreglo $A = [-3, -1, 1, 3, 5]$, el índice $i = 3$ es un punto fijo, ya que $A[3] = 3$.
* En el arreglo $A = [-2, 0, 2, 4, 6]$, el índice $i = 2$ es un punto fijo, pues $A[2] = 2$.
* En cambio, en $A = [-1, 0, 1, 2, 5]$, no existe ningún punto fijo.

---

### 🎯 Tarea

Implemente un algoritmo eficiente que, dado un arreglo $A[0..N]$ **ordenado en forma estrictamente ascendente** (sin valores repetidos), determine si existe un punto fijo.

* Si existe, debe retornar el índice $i$ correspondiente.
* Si no existe, debe retornar $-1$.

---

### 📌 Restricciones

* El arreglo contiene enteros, tanto negativos como positivos.
* El tamaño del arreglo es $1 \leq N \leq 10^6$.
* Se espera que la solución sea lo más eficiente posible. Discuta cuál es la complejidad temporal y espacial de su solución y por qué su solución es la más eficiente.

---

### ✅ Ejemplos

**Entrada 1:**

$$
A = [-3, -1, 1, 3, 5]
$$

**Salida:**

$$
3
$$

**Entrada 2:**

$$
A = [-1, 0, 1, 2, 5]
$$

**Salida:**

$$
-1
$$

```python
from typing import List
#solución ejercicio 2
def e1(data: List[int], left: int = 0, right: int = None) -> int:
  if(right is None):
    right = len(data)-1

  if(left > right):
    return -1

  mid = (left + right) // 2
  if(data[mid] == mid):
    return mid
  elif(data[mid] < mid):
    return e1(data, mid+1, right)
  else:
    return e1(data, left, mid-1)

A=[-3,-1,1,3,5]
e1(l3)
```

```python
from typing import List
#solución ejercicio 2
def e2(data: List[int], x: int, left: int = 0, right: int = None) -> int:
  if(right is None):
    right = len(data)-1

  if(left > right):
    return -1
  print("ventana",data[left:right+1])
  if(left == right):
    size = len(data)
    good_half = left + (size // 2) -1
    if(good_half < size and data[good_half] == x):
      return True
    else:
      return False
  mid = (left + right) // 2
  if(data[mid] >= x):
    return e2(data, x, left, mid)
  else:
    return e2(data, x, mid+1, right)

A = [1, 1, 2, 2, 2, 2,2, 2, 2, 3, 3, 4, 5]
e2(A, 2)
```
