## Problema 1 — Robot de rescate

Una empresa de rescate opera un robot autónomo en una zona de desastre
representada como una cuadrícula de `n` filas y `m` columnas. Cada celda
`(i, j)` tiene asociado un tiempo de tránsito en segundos que tarda el
robot en cruzarla. Las celdas con valor `-1` están bloqueadas por
escombros y el robot no puede atravesarlas.

El robot siempre parte desde la esquina superior izquierda `(0, 0)` y
debe llegar a la esquina inferior derecha `(n-1, m-1)`. Sin embargo,
en el camino **debe pasar obligatoriamente por un puesto de comando**
ubicado en la celda `checkpoint = (x, y)`, donde recibirá instrucciones
actualizadas antes de continuar.

Por restricciones del terreno, **el robot solo puede moverse hacia la
derecha o hacia abajo** en cada paso.

Se define el **costo de una ruta** como la suma de los tiempos de
tránsito de todas las celdas visitadas, incluyendo la celda de inicio,
el checkpoint y la celda destino.

Implementa una función que reciba la cuadrícula y el checkpoint, y
retorne el **costo mínimo** de la ruta completa. Si no existe ninguna
ruta válida que pase por el checkpoint, retorna `-1`.
