**Contexto.** Dos estaciones meteorológicas vecinas registran la temperatura del aire (en décimas de grado) a lo largo del día, pero no miden a las mismas horas ni la misma cantidad de veces, y cada sensor tiene un pequeño error. Para validar que ambos sensores cuentan la misma historia, el equipo técnico quiere encontrar el **mayor número de lecturas que se pueden emparejar en orden**.

Se dice que una lectura `a` de la primera estación y una lectura `b` de la segunda **coinciden** si `|a − b| ≤ tolerancia`. Un emparejamiento escoge lecturas de la primera serie y lecturas de la segunda, la misma cantidad de cada una y respetando el orden en que fueron tomadas, de modo que la primera lectura escogida de una serie coincida con la primera escogida de la otra, la segunda con la segunda, y así sucesivamente. Una lectura se usa a lo sumo una vez.

**Entrada:** las listas `serie_a` y `serie_b` y el entero `tolerancia`.

## Punto 2 — Ruta del dron

**Contexto.** Una empresa de mensajería usa drones que vuelan sobre una ciudad representada como una cuadrícula de `n` filas y `m` columnas. Cada celda tiene un **costo de energía** (un entero positivo) que el dron gasta al pasar por ella; las celdas con valor `-1` son zonas restringidas por las que el dron no puede pasar.

El dron sale de la celda `origen = (x1, y1)` y debe entregar un paquete en la celda `destino = (x2, y2)`, donde la primera coordenada es la fila y la segunda la columna (la fila `0` es la de arriba y la columna `0` la de la izquierda). Por la dirección del viento, en cada paso el dron solo puede moverse de tres maneras:

- a la **derecha**: de `(x, y)` a `(x, y + 1)`,
- hacia **abajo**: de `(x, y)` a `(x + 1, y)`,
- en **diagonal hacia abajo a la derecha**: de `(x, y)` a `(x + 1, y + 1)`. En este movimiento el dron pasa directamente a la celda diagonal, sin pasar por las celdas vecinas.

El **costo de una ruta** es la suma de los costos de todas las celdas por las que pasa, incluyendo el origen y el destino. Determine el **costo mínimo** para ir del origen al destino. Si no existe ninguna ruta posible (por ejemplo, porque el origen o el destino están restringidos, porque las zonas restringidas bloquean el paso o porque el destino no se puede alcanzar con esos movimientos), la respuesta es `-1`.

Por ejemplo, en el mapa

```
 1   3   1   2
 1   5  -1   1
 4   2   1   3
```

para ir de `(0, 0)` a `(2, 3)` la ruta de menor costo es `(0,0) → (1,0) → (2,1) → (2,2) → (2,3)` (abajo, diagonal, derecha, derecha), que cuesta `1 + 1 + 2 + 1 + 3 = 8`. La ruta que va en diagonal siempre que puede, `(0,0) → (1,1) → (2,2) → (2,3)`, pasa por menos celdas pero cuesta `10`. Note que la celda `(1, 2)` está restringida, y que desde `(0, 0)` no se puede llegar, por ejemplo, a una celda de una fila o columna anterior.

**Entrada:** la cuadrícula `mapa` (lista de filas), la tupla `origen` y la tupla `destino`.

**Salida:** el costo mínimo de una ruta, o `-1` si no existe.

**Restricciones:** `1 ≤ n, m ≤ 200`; cada celda vale `-1` o un entero entre `1` y `100`; `origen` y `destino` están dentro de la cuadrícula y pueden estar en cualquier posición (incluso ser la misma celda).
