## Problema 1 — Navegación en almacén inteligente

Una empresa de logística opera un almacén automatizado dividido en una cuadrícula
de `n` filas y `m` columnas. Cada celda `(i, j)` tiene asociado un tiempo de
tránsito en segundos que tarda un robot en cruzarla. Algunas celdas están
bloqueadas por estanterías fijas y se representan con `-1`; el robot no puede
ingresar a ellas.

El robot parte desde la celda `p1 = (f1, c1)` y debe llegar a `p2 = (f2, c2)`.
Por diseño del sistema de rieles, **el robot solo puede moverse hacia la derecha
o hacia abajo** en cada paso.

Se define el **costo de una ruta** como la suma de los tiempos de tránsito de
todas las celdas visitadas, incluyendo la celda de inicio y la celda de destino.

Implementa una función que reciba la cuadrícula, `p1` y `p2`, y retorne el
**costo mínimo** para llegar de `p1` a `p2`. Si no existe ninguna ruta válida
(por obstáculos o por la dirección de movimiento), retorna `-1`.

p1 = (0, 0)

````python
import random
random.seed(1)
def house_robber(nums):
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    
    dp = [0] * len(nums)
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])
    
    for i in range(2, len(nums)):
        dp[i] = max(dp[i-1], nums[i] + dp[i-2])
    
    return dp[-1]


test_cases = []
for i in range(10):
    houses = [random.randint(0, random.randint(1,10000)) for _ in range(random.randint(100, 20000))]
    max_robbed = house_robber(houses)
    test_cases.append((houses, max_robbed))
    
for tc in test_cases:
    print(f"Houses: {tc[0][:10]}... Total Houses: {len(tc[0])}, Max Robbed: {tc[1]}")
````

````python
from typing import Any, Dict, List
from collections import deque

class GraphWL:
  def __init__(self):
    self.adj_list: Dict[Any, List[Any]] = {}
    self.size: int = 0

  def add_vertex(self, value: Any) -> None:
    if value in self.adj_list:
      return None  # ya está el nodo
    self.adj_list[value] = []
    self.size += 1

  def add_edge(self, vertex_1: Any, vertex_2: Any, directed: bool = True, weight: Any = None):
    if vertex_1 not in self.adj_list:
      self.add_vertex(vertex_1)
    if vertex_2 not in self.adj_list:
      self.add_vertex(vertex_2)

    # agregar la arista
    if vertex_2 not in self.adj_list[vertex_1]:
      self.adj_list[vertex_1].append((vertex_2, weight))

    # si no es dirigido, agregar también la inversa
    if not directed and vertex_1 not in self.adj_list[vertex_2]:
      self.adj_list[vertex_2].append((vertex_1, weight))

  def __repr__(self):
    result = ""
    for vertex, neighbors in self.adj_list.items():
      result += f"{vertex}: {neighbors}\n"
    return result
````

````python
def rutas_abastecimiento(G: GraphWL, N: str, K: int, A: int) -> list[list[str]]:
    if N not in G.adj_list:
        return []

    resultados: list[list[str]] = []

    def dfs(actual: str, ruta: list[str], costo: int, aristas: int) -> None:
        if actual == N and aristas > 0:
            if costo > K:
                resultados.append(list(ruta))
            return

        if aristas >= A:
            return

        for vecino, peso in G.adj_list[actual]:
            if vecino not in ruta:
                ruta.append(vecino)
                dfs(vecino, ruta, costo + peso, aristas + 1)
                ruta.pop()

    for origen in G.adj_list:
        dfs(origen, [origen], 0, 0)

    return resultados
````

````python
## Ejercicios final
from collections import Counter

def count_anagram_substrings(s1, s2):
    k = len(s2)
    n = len(s1)
    count_s1 = Counter(s1[:k])
    count_s2 = Counter(s2)
    anagram_count = 0
    
    for i in range(k, n):
        if count_s1 == count_s2:
            anagram_count += 1
        count_s1[s1[i]] += 1
        count_s1[s1[i-k]] -= 1
        if count_s1[s1[i-k]] == 0:
            del count_s1[s1[i-k]]
    
    if count_s1 == count_s2:
        anagram_count += 1
            
    return anagram_count
````

````python
def rectangle_area(a1,a2,b1,b2,c1,c2,d1,d2):
    area_1 = (b1-a1) * (b2-a2)
    area_2 = (c1-d1) * (c2-d2)
    inter_area = (min(d1,b1) - max(a1, c1)) * (min(b2, d2) - max(a2, c2)) 
    print(area_1, area_2, inter_area)
    return area_1 + area_2 - inter_area

rectangle_area(-3,0,3,4,0,-1,9,2)
````

````python
def computeArea(ax1: int, ay1: int, ax2: int, ay2: int, bx1: int, by1: int, bx2: int, by2: int) -> int:
    
    area_1 = (ax2 - ax1) * (ay2 - ay1)               # [1] area of the first rectangle
    area_2 = (bx2 - bx1) * (by2 - by1)               # [2] area of the second rectangle 
    x_over = max(min(ax2,bx2) - max(ax1,bx1),0)      # [3] overlap of sides along x-axis
    y_over = max(min(ay2,by2) - max(ay1,by1),0)      # [4] overlap of sides along y-axis
    area_0 = x_over * y_over                         # [5] area of the overlap region
    print(area_1, area_2, area_0)
    return area_1 + area_2 - area_0

computeArea(-3,0,3,4,0,-1,9,2)
````

````python
from typing import List
class Solution:
    def canChoose(self, groups: List[List[int]], nums: List[int]) -> bool:
        i: int = 0
        j: int = 0
        k: int = 0
        g_count: int = 0
        while(i < len(groups)):
            if(groups[i][j] == nums[k]):
                if(j == len(groups[i])-1):
                    i += 1
                    j = 0
                    k += 1
                    g_count += 1
                else:
                    j += 1
                    k += 1
            else:
                j = 0
                k += 1
            if(k == len(nums)): break
        return g_count == len(groups)

s = Solution()
s.canChoose(groups = [[1,2,3],[3,4]], nums = [7,7,1,2,3,4,7,7])

            
````
