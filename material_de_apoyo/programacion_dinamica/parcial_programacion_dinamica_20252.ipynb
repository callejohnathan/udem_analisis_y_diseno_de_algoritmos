# Ejercicio 1



## 🧩 **Conteo de Equipos de Soldados**

## 🧠 Descripción del problema

Hay `n` soldados formados en una línea.
Cada soldado tiene un valor de **calificación (rating)** único, representado por el arreglo `rating`.

Tu tarea es **formar equipos de 3 soldados** `(i, j, k)` siguiendo las reglas:

* Los índices deben cumplir `0 ≤ i < j < k < n`.
* El equipo es **válido** si los valores de calificación siguen un orden **estrictamente creciente o decreciente**, es decir:

  * `rating[i] < rating[j] < rating[k]`, o
  * `rating[i] > rating[j] > rating[k]`.

Devuelve el **número total de equipos válidos** que se pueden formar.

> ⚠️ Un mismo soldado puede pertenecer a varios equipos.

---

## 📥 **Entrada**

* `rating`: lista de enteros positivos, donde cada elemento representa la calificación de un soldado.

## 📤 **Salida**

* Un número entero que indica cuántos equipos de 3 soldados cumplen las condiciones de orden.

---

## 📋 **Restricciones**

* `n == len(rating)`
* `3 ≤ n ≤ 1000`
* `1 ≤ rating[i] ≤ 10⁵`
* Todos los valores de `rating` son **únicos**.

---

## 🧪 **Casos de prueba**

### Ejemplo 1

```python
rating = [2, 5, 3, 4, 1]
# Resultado esperado = 3
# Equipos válidos: (2,3,4), (5,4,1), (5,3,1)
```

---

### Ejemplo 2

```python
rating = [2, 1, 3]
# Resultado esperado = 0
# No se pueden formar equipos crecientes ni decrecientes.
```

---

### Ejemplo 3

```python
rating = [1, 2, 3, 4]
# Resultado esperado = 4
# Equipos válidos (crecientes): (1,2,3), (1,2,4), (1,3,4), (2,3,4)
```

---

  * Un equipo creciente si existen soldados `i < j` con `rating[i] < rating[j]`
    y soldados `k > j` con `rating[j] < rating[k]`.
  * Un equipo decreciente si existen soldados `i < j` con `rating[i] > rating[j]`
    y soldados `k > j` con `rating[j] > rating[k]`.
