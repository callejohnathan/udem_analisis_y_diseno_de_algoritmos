## 1️⃣ ¿Qué es la fuerza bruta?

### 📌 Definición

> **Fuerza bruta** es una estrategia algorítmica que consiste en **explorar exhaustivamente todas las soluciones posibles** de un problema, sin aplicar optimizaciones ni heurísticas, y seleccionar aquellas que cumplen la condición requerida.

### 🧠 Ideas clave

* No asume estructura del problema.
* No descarta soluciones anticipadamente.
* **Garantiza corrección**, siempre que haya tiempo suficiente.
* Es conceptualmente simple, pero computacionalmente costosa.

🔎 Frase para enfatizar:

> *“La fuerza bruta no es mala programación; es el punto de partida honesto para entender un problema.”*

---

## 2️⃣ Fuerza bruta como estrategia algorítmica

### 🔍 Características

✔️ Explora todo el espacio de búsqueda
✔️ Fácil de implementar
✔️ Difícil de escalar
✔️ Excelente para:

* Validar soluciones óptimas
* Casos pequeños
* Prototipos

### ⚠️ Principal problema

El **crecimiento del número de casos** a evaluar suele ser:

* Cuadrático
* Exponencial
* Factorial

Y esto **no escala**.

---

## 3️⃣ Introducción al análisis de complejidad

### 🤔 Pregunta fundamental

> ¿Qué pasa cuando el tamaño de la entrada crece?

El análisis de complejidad busca responder:

* ¿Cuántas operaciones ejecuta un algoritmo?
* ¿Cómo crece ese número respecto a `n`?

---

### 📐 Complejidad temporal

Mide **cuántas operaciones** ejecuta un algoritmo.

Ejemplos intuitivos:

* Un ciclo → depende de `n`
* Dos ciclos anidados → depende de `n²`
* Generar subconjuntos → depende de `2ⁿ`

---

### 🧮 Notación Big-O (idea básica)

> Big-O describe **el comportamiento asintótico** de un algoritmo cuando `n` crece.

| Notación | Interpretación |
| -------- | -------------- |
| O(1)     | Constante      |
| O(n)     | Lineal         |
| O(n²)    | Cuadrática     |
| O(2ⁿ)    | Exponencial    |
| O(n!)    | Factorial      |

📌 **No mide tiempo real**, mide **cómo crece el tiempo**.

---

## 4️⃣ Crecimiento asintótico (el golpe de realidad)

### 📈 Ejemplo comparativo

Supongamos que una operación tarda 1 microsegundo:

| n  | O(n)  | O(n²)  | O(2ⁿ)  |
| -- | ----- | ------ | ------ |
| 10 | 10 µs | 100 µs | 1 ms   |
| 20 | 20 µs | 400 µs | 1 s    |
| 30 | 30 µs | 900 µs | 17 min |

🔴 **Conclusión**:

> Un algoritmo puede ser correcto… y aun así ser inutilizable.

---

## 5️⃣ ¿Cuándo la fuerza bruta es aceptable?

### ✅ Casos donde **sí** tiene sentido

* Tamaños pequeños de entrada
* Validación de resultados
* Problemas educativos
* Comparar con versiones optimizadas

### 🚨 Señales de alerta (code smells algorítmicos)

* Múltiples ciclos anidados
* Enumeración de combinaciones o permutaciones
* Repetición de cálculos
* Estados que se vuelven inválidos temprano pero siguen explorándose

💡 Estas señales abren la puerta a:

* Backtracking
* Programación dinámica
* Algoritmos greedy

## 🧩 Problema 1: Suma de pares

**Enunciado**
Dado un arreglo de `n` enteros y un valor objetivo `x`, determine si existe algún par de elementos cuya suma sea igual a `x`.

📌 **Restricción**:
Use una solución de fuerza bruta.

💭 **Preguntas guía**:

* ¿Cuántos pares se evalúan?
* ¿Cómo crece el número de comparaciones?





## ✅ ¿Cuántos pares se evalúan? 🔎

Si el arreglo tiene (n) elementos, la cantidad de pares distintos ((i,j)) con (i<j) es:
$
[
\binom{n}{2} = \frac{n(n-1)}{2}
]
$
📌 Ejemplo rápido:

$
* si (n=5) → (\frac{5\cdot4}{2} = 10) pares.
$
---

## 📈 ¿Cómo crece el número de comparaciones? 📊

En fuerza bruta, en el peor caso (cuando **no existe** el par, o existe y está al final), se comparan todos los pares:
$
[
\frac{n(n-1)}{2} \in \Theta(n^2)
]
$

✅ **Complejidad temporal:** (O(n^2))
✅ **Complejidad espacial:** (O(1))

---

## 🧠 Solución fuerza bruta (sin librerías, con typehints) ✅

```python
def existe_par_suma_x(arr: list[int], x: int) -> bool:
    n: int = len(arr)

    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == x:
                return True

    return False
```

---

## 🧪 Casos de prueba rápidos

```python
def _probar() -> None:
    print(existe_par_suma_x([2, 7, 11, 15], 9))    # True (2 + 7)
    print(existe_par_suma_x([1, 2, 3, 4], 8))      # False
    print(existe_par_suma_x([-3, 1, 4, 2], -1))    # True (-3 + 2)

_probar()
```

---

## 🎯 Extra didáctico

* El algoritmo **no usa memoria extra**: solo índices.
* La idea clave de fuerza bruta aquí es: **“si no tengo estrategia, lo intento todo”**.
* Es un buen punto de partida para luego contrastar con:

  * **hashing / diccionarios** (O(n)) (no fuerza bruta)
  * **ordenar + dos punteros** (O(n \log n))

## 🧩 Problema 2: Subconjunto con suma objetivo

**Enunciado**
Dado un conjunto de `n` números enteros positivos y un valor `S`, determine si existe algún subconjunto cuya suma sea exactamente `S`.

📌 **Restricción**:
Explore todos los subconjuntos posibles.

💭 **Preguntas guía**:

* ¿Cuántos subconjuntos existen?
* ¿Qué pasa si `n` aumenta ligeramente?




# 🧠 Análisis Conceptual

## 🔢 ¿Cuántos subconjuntos existen?

Si tenemos un conjunto de tamaño `n`, cada elemento tiene 2 opciones:

* 🔹 Se incluye
* 🔹 No se incluye

Por lo tanto:

$$
\text{Número total de subconjuntos} = 2^n
$$

Incluyendo:

* El conjunto vacío
* El conjunto completo

---

### 📊 Ejemplo pequeño

Si `n = 3` → {a, b, c}

Subconjuntos:

* {}
* {a}
* {b}
* {c}
* {a,b}
* {a,c}
* {b,c}
* {a,b,c}

Total:

$$
2^3 = 8
$$

---

## 📈 ¿Qué pasa si n aumenta ligeramente?

Aquí es donde empieza el caos 😈

| n  | Subconjuntos      |
| -- | ----------------- |
| 10 | 1,024             |
| 20 | 1,048,576         |
| 30 | 1,073,741,824     |
| 40 | 1,099,511,627,776 |

💥 El crecimiento es **exponencial**.

Formalmente:

$$
T(n) \in O(2^n)
$$

Este tipo de crecimiento es muchísimo más agresivo que (O(n^2)).

---

# 🚀 Implementación Fuerza Bruta (sin librerías, con typehints)

Vamos a generar los subconjuntos usando la idea clásica de **máscaras binarias**, pero implementadas manualmente con ciclos.

```python
def existe_subconjunto_suma(nums: list[int], S: int) -> bool:
    n: int = len(nums)

    total_subconjuntos: int = 1 << n  # 2^n

    for mascara in range(total_subconjuntos):
        suma_actual: int = 0

        for i in range(n):
            if mascara & (1 << i):
                suma_actual += nums[i]

        if suma_actual == S:
            return True

    return False
```

---

# 🔍 Explicación didáctica

Si `n = 3`, las máscaras van de:

```
000
001
010
011
100
101
110
111
```

Cada bit indica si el elemento se incluye o no.

Es decir:

```
101  → incluye nums[0] y nums[2]
```

---

# 📌 Complejidad

## ⏳ Tiempo

Tenemos:

* (2^n) subconjuntos
* En cada subconjunto revisamos hasta (n) elementos

Por lo tanto:

$$
T(n) = n \cdot 2^n
$$

En notación asintótica:

$$
O(n 2^n)
$$

---

## 🧠 Espacio

Solo usamos variables auxiliares:

$$
O(1)
$$


---


# ✅ Versión iterativa (generando subconjuntos paso a paso)

## 💡 Idea

Vamos construyendo una lista de sumas posibles de subconjuntos:

* Arrancamos con el subconjunto vacío → suma `0`
* Por cada número `v`, duplicamos lo que tenemos:

  * dejamos las sumas existentes (no tomar `v`)
  * agregamos nuevas sumas (tomar `v`): `s + v`

Esto genera efectivamente los (2^n) subconjuntos, pero de forma iterativa.

---

## 🧩 Implementación (sin librerías, con typehints)

```python
def existe_subconjunto_suma_iterativo(nums: list[int], S: int) -> bool:
    # sumas_subconjuntos guarda las sumas de TODOS los subconjuntos generados hasta el momento
    sumas_subconjuntos: list[int] = [0]

    for v in nums:
        # Guardamos cuántas sumas había antes de procesar v
        tam: int = len(sumas_subconjuntos)

        # Para cada suma existente, generamos una nueva sumando v (esto crea la "mitad" nueva)
        for i in range(tam):
            nueva_suma: int = sumas_subconjuntos[i] + v

            if nueva_suma == S:
                return True

            sumas_subconjuntos.append(nueva_suma)

    # Ojo: si S == 0, el subconjunto vacío ya cumple
    return S == 0
```

---

# 🔎 ¿Por qué esto es fuerza bruta?

Porque el tamaño de `sumas_subconjuntos` se duplica por cada elemento:

* después de 1 número → 2 sumas
* después de 2 números → 4 sumas
* …
* después de n números → (2^n) sumas

Es decir, **estamos generando todos los subconjuntos**, solo que “implícitamente” mediante sus sumas.

---

# 📈 Complejidad

* Tiempo: en total se agregan (2^n - 1) sumas → **(O(2^n))** operaciones dominantes (más el overhead de lista, pero asintóticamente es exponencial).
* Espacio: guardamos hasta (2^n) sumas → **(O(2^n))**.

📌 Comparación didáctica:

* Bitmask: (O(n2^n)) tiempo, (O(1)) espacio.
* Construcción iterativa: ~(O(2^n)) tiempo, **(O(2^n)) espacio**.

---

# 🧪 Casos de prueba

```python
def _probar() -> None:
    print(existe_subconjunto_suma_iterativo([3, 34, 4, 12, 5, 2], 9))   # True (4+5)
    print(existe_subconjunto_suma_iterativo([1, 2, 3], 7))              # False
    print(existe_subconjunto_suma_iterativo([8, 1, 2], 0))              # True (vacío)

_probar()
```

# ✅ Versión iterativa que muestra los subconjuntos generados 🧾

## 🧠 Idea

Mantenemos una lista de subconjuntos ya construidos.
Por cada valor `v`, duplicamos la lista:

* Los subconjuntos existentes se quedan tal cual (no tomar `v`)
* Se crean nuevos subconjuntos agregando `v` a cada uno (tomar `v`)

---

## 🧩 Código (imprime cada subconjunto a medida que se crea)

```python
def mostrar_subconjuntos(nums: list[int]) -> None:
    subconjuntos: list[list[int]] = [[]]  # empezamos con el vacío

    print("Subconjunto:", [])  # mostramos el vacío

    for v in nums:
        tam: int = len(subconjuntos)

        # por cada subconjunto existente, creamos uno nuevo agregando v
        for i in range(tam):
            nuevo: list[int] = subconjuntos[i] + [v]
            subconjuntos.append(nuevo)
            print("Subconjunto:", nuevo)
```

---

# ✅ Variante: mostrar subconjunto + suma (útil para el objetivo S) 🎯

```python
def mostrar_subconjuntos_con_suma(nums: list[int]) -> None:
    subconjuntos: list[list[int]] = [[]]

    print("Subconjunto:", [], "Suma:", 0)

    for v in nums:
        tam: int = len(subconjuntos)

        for i in range(tam):
            nuevo: list[int] = subconjuntos[i] + [v]
            subconjuntos.append(nuevo)

            suma_nuevo: int = 0
            for x in nuevo:
                suma_nuevo += x

            print("Subconjunto:", nuevo, "Suma:", suma_nuevo)
```

📌 Aquí la suma la calculo con ciclo (sin `sum()`), para mantenerlo “puro” y didáctico.

---

# 🧪 Ejemplo rápido

```python
def _probar() -> None:
    nums: list[int] = [1, 2, 3]
    mostrar_subconjuntos(nums)

_probar()
```

Salida esperada (orden típico):

* []
* [1]
* [2]
* [1, 2]
* [3]
* [1, 3]
* [2, 3]
* [1, 2, 3]

## 🧩 Problema 3: Permutaciones válidas

**Enunciado**
Dada una cadena de longitud `n`, genere todas sus permutaciones y determine si alguna cumple una condición específica (por ejemplo, ser palíndroma).

📌 **Restricción**:
No descarte permutaciones anticipadamente.

💭 **Preguntas guía**:

* ¿Por qué `n!` crece tan rápido?
* ¿En qué punto el problema se vuelve inviable?



---

# 🧠 Preguntas guía (explicadas)

## 1️⃣ ¿Por qué (n!) crece tan rápido? 🚀

Porque al permutar:

* para el primer carácter hay `n` opciones,
* para el segundo quedan `n-1`,
* luego `n-2`,
* …
* hasta `1`.

Entonces:

[
n! = n \cdot (n-1) \cdot (n-2) \cdots 1
]

📌 La multiplicación acumulativa hace que el crecimiento sea **explosivo** (mucho más que (2^n) y que (n^2)).

---

## 2️⃣ ¿En qué punto el problema se vuelve inviable? ⚠️

Depende del costo por permutación.

* Generar una permutación cuesta al menos (O(n)) (armar/guardar/imprimir).
* Verificar palíndromo cuesta (O(n)).

➡️ En total típico: **(O(n)) por permutación**
y hay **(n!)** permutaciones.

[
T(n) \in O(n \cdot n!)
]

### 📌 Números para sentirlo (aprox.)

* (10! = 3,628,800) → todavía “posible” en un equipo decente si no imprimes todo.
* (11! \approx 39,916,800) → ya pesado.
* (12! \approx 479,001,600) → empieza a ser muy duro.
* (13! \approx 6,227,020,800) → inviable para clase/laptop.

✅ En práctica, **imprimir** todas las permutaciones vuelve inviable el ejercicio mucho antes.

---

# ✅ Solución iterativa (sin librerías) + condición palíndroma

## 🎯 Estrategia iterativa

Usaremos el algoritmo de **“siguiente permutación” (next permutation)**:

1. Ordenamos los caracteres (para empezar por la menor permutación).
2. Repetimos:

   * evaluamos la permutación actual,
   * generamos la siguiente en orden lexicográfico,
   * hasta que no exista más.

📌 Esto genera **todas** sin recursión y sin librerías.

---

## 🧩 Código completo

```python
def es_palindroma(chars: list[str]) -> bool:
    i: int = 0
    j: int = len(chars) - 1

    while i < j:
        if chars[i] != chars[j]:
            return False
        i += 1
        j -= 1

    return True


def ordenar_insercion(chars: list[str]) -> None:
    # Insertion sort (sin librerías)
    n: int = len(chars)
    for i in range(1, n):
        key: str = chars[i]
        j: int = i - 1
        while j >= 0 and chars[j] > key:
            chars[j + 1] = chars[j]
            j -= 1
        chars[j + 1] = key


def next_permutation(a: list[str]) -> bool:
    """
    Transforma 'a' en la siguiente permutación lexicográfica.
    Retorna False si 'a' ya es la última permutación (orden descendente).
    """
    n: int = len(a)

    # 1) encontrar el pivote: el mayor i tal que a[i] < a[i+1]
    i: int = n - 2
    while i >= 0 and a[i] >= a[i + 1]:
        i -= 1
    if i < 0:
        return False  # ya no hay siguiente

    # 2) encontrar el mayor j > i tal que a[j] > a[i]
    j: int = n - 1
    while a[j] <= a[i]:
        j -= 1

    # 3) swap a[i], a[j]
    tmp: str = a[i]
    a[i] = a[j]
    a[j] = tmp

    # 4) revertir el sufijo a[i+1:]
    l: int = i + 1
    r: int = n - 1
    while l < r:
        tmp2: str = a[l]
        a[l] = a[r]
        a[r] = tmp2
        l += 1
        r -= 1

    return True


def existe_permutacion_palindroma(s: str) -> bool:
    chars: list[str] = []
    for c in s:
        chars.append(c)

    # Empezar desde la primera permutación (ordenada)
    ordenar_insercion(chars)

    # Evaluar la primera y luego todas las siguientes
    if es_palindroma(chars):
        return True

    while next_permutation(chars):
        if es_palindroma(chars):
            return True

    return False
```

---

# 🧪 Casos de prueba

```python
def _probar() -> None:
    print(existe_permutacion_palindroma("aab"))   # True -> "aba"
    print(existe_permutacion_palindroma("abc"))   # False
    print(existe_permutacion_palindroma("aabb"))  # True -> "abba" / "baab"

_probar()
```

---

# 🧠 Comentarios didácticos (para rematar la clase) 🧠

* Aquí **sí** hacemos fuerza bruta: generamos permutación por permutación y revisamos.
* El cuello de botella es que el espacio de búsqueda es **factorial**: (n!).
* Aunque “solo” busquemos una que cumpla, **en peor caso** toca recorrerlas todas.
