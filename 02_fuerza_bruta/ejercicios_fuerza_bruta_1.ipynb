# 🧩 Problema 1: Subarreglo de suma máxima (versión fuerza bruta)

## 📘 Enunciado

Dado un arreglo de `n` enteros (positivos y negativos), determine el subarreglo contiguo cuya suma sea máxima.

📌 **Restricción:** Evalúe todos los subarreglos posibles.

---

## 💭 Preguntas guía

* ¿Cuántos subarreglos contiguos existen en un arreglo de tamaño `n`?
* ¿Qué complejidad tiene evaluar cada uno?
* ¿Es lo mismo que explorar todos los subconjuntos?
* ¿Cómo se compara contra el algoritmo de Kadane?

---

## 🧠 Clave conceptual

Número de subarreglos contiguos:
$
[
\frac{n(n+1)}{2}
]
$
Complejidad típica fuerza bruta:
$
[
O(n^3) \quad (\text{si se recalcula la suma cada vez})
]
$
Optimizable a:
$
[
O(n^2)
]
$

# 🧩 Problema 2: Tripleta con suma objetivo

## 📘 Enunciado

Dado un arreglo de `n` enteros y un valor `X`, determine si existen tres elementos distintos cuya suma sea exactamente `X`.

📌 **Restricción:** Evalúe todas las posibles tripletas.

---

## 💭 Preguntas guía

* ¿Cuántas tripletas distintas existen?
* ¿Cuál es la relación con combinatoria?
* ¿Qué orden de complejidad se obtiene?
* ¿Cómo cambiaría si fueran 4 elementos? ¿k elementos?

---

## 🧠 Clave conceptual

Número de tripletas:
$
[
\binom{n}{3} = \frac{n(n-1)(n-2)}{6}
]
$
Complejidad:
$
[
O(n^3)
]
$
