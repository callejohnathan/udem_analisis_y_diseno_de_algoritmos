## 📘 Heaps en Python (Min-Heap y Max-Heap)

## 1️⃣ Min-Heap (comportamiento por defecto)

Python implementa por defecto un **min-heap** usando `heapq`.

Propiedad:

[
$heap[parent] \le heap[child]$
]

### Crear heap

```python
import heapq
from typing import List

nums: List[int] = [5, 1, 8, 3]

heapq.heapify(nums)

print(nums)
```

Complejidad:

```
O(n)
```

---

### Insertar elemento

```python
heapq.heappush(nums, 2)
```

Complejidad:

```
O(log n)
```

---

### Extraer mínimo

```python
smallest: int = heapq.heappop(nums)
```

Complejidad:

```
O(log n)
```

---

### Consultar mínimo

```python
nums[0]
```

Complejidad:

```
O(1)
```

---

# 2️⃣ Max-Heap (nuevo soporte en Python moderno)

Desde **Python 3.14** `heapq` incluye funciones específicas para max-heaps.

Estas funciones son:

```
heapq.heapify_max()
heapq.heappush_max()
heapq.heappop_max()
heapq.heapreplace_max()
heapq.heappushpop_max()
```

---

## Crear un Max-Heap

```python
import heapq
from typing import List

nums: List[int] = [5, 1, 8, 3]

heapq.heapify_max(nums)

print(nums)
```

Ahora el elemento máximo está en la raíz.

```
nums[0]  # mayor elemento
```

---

## Insertar en Max-Heap

```python
heapq.heappush_max(nums, 10)
```

Complejidad:

```
O(log n)
```

---

## Extraer máximo

```python
largest: int = heapq.heappop_max(nums)
```

Complejidad:

```
O(log n)
```

---

# 3️⃣ Comparación rápida

| Operación      | Min-Heap     | Max-Heap         |
| -------------- | ------------ | ---------------- |
| construir heap | `heapify()`  | `heapify_max()`  |
| insertar       | `heappush()` | `heappush_max()` |
| extraer        | `heappop()`  | `heappop_max()`  |
| raíz           | mínimo       | máximo           |

---

# 4️⃣ Compatibilidad (muy importante para estudiantes)

Muchos entornos todavía usan **Python < 3.14**, donde **no existe max-heap nativo**.

Por eso es común ver el patrón:

```python
heapq.heappush(heap, -value)
```

y luego:

```python
value = -heapq.heappop(heap)
```

Este patrón sigue siendo **válido y portable**.

---

# 5️⃣ Complejidad de operaciones

| Operación | Complejidad |
| --------- | ----------- |
| `heapify` | O(n)        |
| `push`    | O(log n)    |
| `pop`     | O(log n)    |
| ver raíz  | O(1)        |

---

# 6️⃣ Cuándo usar heaps

Los heaps son útiles cuando repetidamente necesitas:

```
extraer el mínimo / máximo
```

Ejemplos clásicos:

* **Dijkstra**
* **Prim**
* **Huffman coding**
* **Connecting ropes**
* **Top-K elements**
* **Scheduling**

---

💡 **Tip pedagógico**

> Un heap **no ordena todo**, solo garantiza que **el mejor elemento está arriba**.

## Complejidad de las operaciones en un Heap

## 1️⃣ Altura del heap (la clave de todo)

Un **heap binario** es un **árbol binario completo**.

Eso significa que su altura es:

[
h = \log_2 n
]

porque cada nivel aproximadamente **duplica el número de nodos**.

Ejemplo:

```
nivel 0      1 nodo
nivel 1      2 nodos
nivel 2      4 nodos
nivel 3      8 nodos
```

Entonces:

[
$n \approx 2^h$
]

y por lo tanto:

[
$h = \log n$
]

⚠️ Esto es lo que explica por qué muchas operaciones cuestan **O(log n)**.

---

# 2️⃣ Insertar elemento (push)

### Operación

```python
heapq.heappush(heap, x)
```

### Qué ocurre internamente

1. El nuevo elemento se agrega **al final del arreglo**.
2. Luego se realiza **sift-up** (bubble-up):

El elemento sube en el árbol mientras sea menor que su padre.

Ejemplo:

```
        2
      /   \
     5     4
    /
   8
```

Insertamos `1`:

```
        2
      /   \
     5     4
    / \
   8   1
```

Luego **sube**:

```
        2
      /   \
     1     4
    / \
   8   5
```

Luego vuelve a subir:

```
        1
      /   \
     2     4
    / \
   8   5
```

### Complejidad

El elemento puede subir **como máximo hasta la raíz**.

Número máximo de pasos:

[
$\log n$
]

Entonces:

```
O(log n)
```

---

# 3️⃣ Extraer mínimo / máximo (pop)

### Operación

```python
heapq.heappop(heap)
```

### Qué ocurre internamente

1️⃣ Se elimina la raíz.

2️⃣ El último elemento del arreglo se mueve a la raíz.

3️⃣ Se hace **sift-down** (bubble-down).

Ejemplo:

Heap inicial:

```
        1
      /   \
     2     4
    / \
   8   5
```

Quitamos `1`.

Último nodo sube:

```
        5
      /   \
     2     4
    /
   8
```

Ahora baja:

```
        2
      /   \
     5     4
    /
   8
```

### Complejidad

El elemento puede bajar **hasta la hoja más profunda**.

Altura del árbol:

[
$\log n$
]

Entonces:

```
O(log n)
```

---

# 4️⃣ Consultar mínimo o máximo

### Operación

```python
heap[0]
```

El mínimo (o máximo en max-heap) **siempre está en la raíz**.

No se mueve nada.

Complejidad:

```
O(1)
```

---

# 5️⃣ Construir heap (heapify)

### Operación

```python
heapq.heapify(lista)
```

Esto transforma un arreglo en heap.

Muchos estudiantes creen que cuesta:

```
O(n log n)
```

pero en realidad cuesta:

```
O(n)
```

---

## ¿Por qué es O(n)?

`heapify` funciona haciendo **sift-down desde los nodos internos hacia arriba**.

Observación clave:

* La mayoría de nodos están **cerca de las hojas**.
* Esos nodos tienen **muy poca altura que recorrer**.

Distribución aproximada:

| Nivel          | nodos | costo |
| -------------- | ----- | ----- |
| hojas          | n/2   | 0     |
| nivel anterior | n/4   | 1     |
| siguiente      | n/8   | 2     |

La suma total:

[
$\sum_{i=0}^{\log n} \frac{n}{2^{i+1}} \cdot i$
]

esto converge a:

```
O(n)
```

Este resultado es uno de los **clásicos de análisis de algoritmos**.

---

# 6️⃣ Complejidad resumida

| Operación      | Complejidad |
| -------------- | ----------- |
| ver mínimo     | O(1)        |
| insertar       | O(log n)    |
| extraer mínimo | O(log n)    |
| heapify        | O(n)        |

---

# 7️⃣ Intuición final (la frase que les ayuda a recordarlo)

Un heap cuesta **log n** porque:

> las operaciones solo recorren la **altura del árbol**, no todos los elementos.

# 1) 🪟 Sliding Window — Subarreglo más largo con suma ≤ K

## ✅ Enunciado formal

Dado un arreglo `A` de longitud `n` compuesto por **enteros no negativos** y un entero `K`, encuentre la **longitud máxima** de un **subarreglo contiguo** cuya suma sea **menor o igual que `K`**.

### 📥 Entrada

* `A`: lista de enteros **no negativos**
* `K`: entero

### 📤 Salida

* Un entero: la longitud máxima de un subarreglo contiguo con suma `≤ K`

### 🧪 Ejemplos

**Ejemplo 1**
Entrada: `A = [1, 2, 1, 1, 1, 3, 2]`, `K = 4`
Salida: `4`
Explicación: `[1,1,1,1]` suma 4.

**Ejemplo 2**
Entrada: `A = [2, 2, 2]`, `K = 1`
Salida: `0`
Explicación: Ningún subarreglo cabe.

---

## ✅ Solución (Python + typehints)

```python
from typing import List

def longest_subarray_sum_at_most_k(A: List[int], K: int) -> int:
    """
    Retorna la longitud máxima de un subarreglo contiguo con suma <= K.
    Requiere: A contiene enteros NO NEGATIVOS.
    """
    n: int = len(A)
    left: int = 0
    current_sum: int = 0
    best: int = 0

    for right in range(n):
        current_sum += A[right]

        # Si la suma se pasó, avanzamos left hasta que vuelva a ser válida.
        while left <= right and current_sum > K:
            current_sum -= A[left]
            left += 1

        # En este punto, la ventana [left..right] cumple suma <= K
        # (o left == right+1 si K < 0, pero K suele ser >= 0 en estos problemas)
        window_len: int = right - left + 1
        if window_len > best:
            best = window_len

    return best
```

---

## 🧠 Análisis greedy (por qué Sliding Window funciona aquí)

### ✅ La idea greedy

El “movimiento greedy” es:

* **Expandir la ventana a la derecha** siempre que sea posible.
* Si se viola la restricción (`suma > K`), **recortar por la izquierda** lo mínimo necesario hasta volver a cumplir.

Es greedy porque en cada paso tomas una decisión local:

* “Si meter un elemento no rompe la regla, me conviene meterlo porque aumenta la longitud”.

### 🔑 Propiedad clave: Monotonía (NO NEGATIVOS)

Este enfoque es correcto **porque todos los elementos son no negativos**:

* Al mover `right` hacia la derecha, la suma **no disminuye**.
* Al mover `left` hacia la derecha, la suma **no aumenta**.

Eso hace que el proceso de “ajustar” sea seguro: si te pasaste, **solo tiene sentido** reducir desde la izquierda.

### 😈 Devils advocate (cuándo NO es válido)

Si el arreglo tuviera números negativos, la suma **ya no es monotónica**, y esta estrategia puede fallar (porque agregar un negativo podría “arreglar” una ventana que antes se pasaba, y el algoritmo no lo anticipa de forma segura).

📌 Conclusión: **Sliding Window estándar para suma ≤ K requiere no negativos.**

---

## ⏱️ Complejidad

* **Tiempo:** `O(n)`
  Cada índice (`left` y `right`) avanza como máximo `n` veces. Aunque haya `while`, `left` no retrocede.
* **Espacio:** `O(1)`
  Solo usamos variables auxiliares.

---

## ✅ Casos de prueba rápidos

```python
def test_longest_subarray_sum_at_most_k() -> None:
    assert longest_subarray_sum_at_most_k([1, 2, 1, 1, 1, 3, 2], 4) == 4
    assert longest_subarray_sum_at_most_k([2, 2, 2], 1) == 0
    assert longest_subarray_sum_at_most_k([], 10) == 0
    assert longest_subarray_sum_at_most_k([0, 0, 0], 0) == 3
    assert longest_subarray_sum_at_most_k([3, 1, 1, 1], 3) == 3  # [1,1,1]

test_longest_subarray_sum_at_most_k()
print("OK ✅")
```

# 2) 🧺 Heaps — Minimizar costo al unir cuerdas

## ✅ Enunciado formal

Tienes `n` cuerdas con longitudes `A[0..n-1]`.
Unir dos cuerdas de longitudes `x` e `y` cuesta `x + y` y produce una nueva cuerda de longitud `x + y`.
Debes repetir hasta que quede una sola cuerda.

**Objetivo:** minimizar el **costo total acumulado**.

### 📥 Entrada

* `A`: lista de enteros (longitudes de cuerdas)

### 📤 Salida

* Un entero: costo mínimo total

### 🧪 Ejemplo

Entrada: `A = [4, 3, 2, 6]`
Salida: `29`
Explicación:

* 2+3=5 (costo 5)
* 4+5=9 (costo 9)
* 6+9=15 (costo 15)
  Total = 5+9+15 = 29

---

## ✅ Solución (Python + typehints) usando Min-Heap

```python
from typing import List
import heapq

def min_cost_to_connect_ropes(A: List[int]) -> int:
    """
    Retorna el costo mínimo total para unir todas las cuerdas.
    Estrategia greedy: siempre unir las dos cuerdas más cortas.
    """
    n: int = len(A)
    if n <= 1:
        return 0

    heap: List[int] = A[:]  # copia para no mutar el input
    heapq.heapify(heap)

    total_cost: int = 0

    while len(heap) > 1:
        x: int = heapq.heappop(heap)
        y: int = heapq.heappop(heap)
        cost: int = x + y
        total_cost += cost
        heapq.heappush(heap, cost)

    return total_cost
```

---

## 🧠 Análisis greedy (por qué “dos más pequeñas” es óptimo)

### ✅ La intuición greedy

Cada vez que unes dos cuerdas, su suma `x+y`:

* se paga **una vez** ahora,
* pero además se convierte en una cuerda nueva que **volverá a participar** en sumas futuras.

Entonces, si haces grande una cuerda “temprano”, ese tamaño grande se **reusa** muchas veces y dispara el costo.

👉 Por eso, la mejor decisión local repetida es:
**unir siempre las dos más pequeñas**, minimizando el “peso futuro” acumulado.

### 🔁 Conexión conceptual

Este problema es el mismo patrón que:

* **Optimal Merge Pattern**
* **Huffman coding** (combinar las dos frecuencias más pequeñas)

### 😈 Devils advocate (qué pasa si uno une “cualquiera”)

Ejemplo: `A = [1, 2, 100, 100]`

* Uniendo grandes primero: costo enorme (se “arrastran” hacia el futuro).
* Uniendo pequeñas primero: costo mucho menor.

La diferencia no es sutil: suele ser **dramática**.

---

## ⏱️ Complejidad

* **Tiempo:** `O(n log n)`

  * `heapify`: `O(n)`
  * Se realizan `n-1` uniones.
    Cada unión hace 2 `pop` + 1 `push`, cada uno `O(log n)` → total `O(n log n)`.
* **Espacio:** `O(n)`
  Por el heap (copia de la lista).

---

## ✅ Casos de prueba rápidos

```python
def test_min_cost_to_connect_ropes() -> None:
    assert min_cost_to_connect_ropes([4, 3, 2, 6]) == 29
    assert min_cost_to_connect_ropes([1, 2, 100, 100]) == 309
    assert min_cost_to_connect_ropes([5]) == 0
    assert min_cost_to_connect_ropes([]) == 0
    assert min_cost_to_connect_ropes([1, 1, 1, 1]) == 8  # (1+1)=2,(1+1)=2,(2+2)=4 => 8

test_min_cost_to_connect_ropes()
print("OK ✅")
```

# 🧩 Cierre de clase: ¿qué hace “greedy” a cada uno?

* 🪟 Sliding Window: greedy por **expansión local** + corrección mínima, apoyado en **monotonía** (no negativos).
* 🧺 Heaps: greedy por elegir repetidamente el **mejor par local** (mínimos), porque reduce el costo “arrastrado” al futuro.

---




````python
import heapq
from typing import List
def min_ropes(ropes: List[int]) -> int:
  heapq.heapify(ropes) #o(n)

  total_cost: int = 0 #o(1)
  while(len(ropes) > 1): #o(nlogn)
    rope_1 = heapq.pop(ropes) #o(log(n))
    rope_2 = heapq.pop(ropes) #o(log(n))

    cost: int = rope_1 + rope_2 #o(1)
    heapq.push(ropes, cost) #o(log(n))
    total_cost += cost #o(1)

  return total_cost

````
