# 🧪 Enunciado: **Coincidencia con comodines**

## 🎯 Objetivo

Dadas dos cadenas:

* `s`: cadena de entrada
* `p`: patrón

Implementa un algoritmo que determine si el patrón **coincide completamente** con la cadena, usando las siguientes reglas de comodines:

* `'?'` coincide con **exactamente un carácter cualquiera**.
* `'*'` coincide con **cualquier secuencia de caracteres** (incluyendo la secuencia vacía).

La coincidencia debe cubrir **toda la cadena** de entrada (no coincidencias parciales).

## 📥 Entrada

* `s`: string con solo letras minúsculas inglesas.
* `p`: string con letras minúsculas inglesas, `'?'` o `'*'`.

## 📤 Salida

* `true` si `p` coincide completamente con `s`.
* `false` en caso contrario.

## 🔒 Restricciones

* `0 ≤ |s|, |p| ≤ 2000`
* Caracteres de `s`: `a-z`
* Caracteres de `p`: `a-z`, `'?'`, `'*'`

## 🧰 Ejemplos

**Ejemplo 1**
Entrada: `s = "aa"`, `p = "a"`
Salida: `false`
Explicación: `"a"` no cubre toda la cadena `"aa"`.

**Ejemplo 2**
Entrada: `s = "aa"`, `p = "*"`
Salida: `true`
Explicación: `'*'` puede representar `"aa"`.

**Ejemplo 3**
Entrada: `s = "cb"`, `p = "?a"`
Salida: `false`
Explicación: `'?'` casa con `'c'`, pero `'a'` no casa con `'b'`.

# 🧪 Enunciado: **Jump Game — ¿puedes llegar al final?**

## 🎯 Objetivo

Se te da un arreglo de enteros `nums`, donde estás inicialmente en el índice `0` y cada `nums[i]` indica **la longitud máxima de salto** que puedes realizar desde `i`.
**Determina si es posible llegar (o sobrepasar) el último índice** del arreglo.

## 📥 Entrada

* `nums`: arreglo de enteros no negativos.

## 📤 Salida

* `true` si puedes llegar al último índice.
* `false` en caso contrario.

## 🔒 Restricciones

* `1 ≤ len(nums) ≤ 10^4`
* `0 ≤ nums[i] ≤ 10^5`

## 🧪 Ejemplos

**Ejemplo 1**
Entrada: `nums = [2,3,1,1,4]`
Salida: `true`
Explicación: desde `0` saltas a `1` (≤2), y desde `1` saltas 3 posiciones hasta el final.

**Ejemplo 2**
Entrada: `nums = [3,2,1,0,4]`
Salida: `false`
Explicación: inevitablemente quedas en `3`, donde `nums[3]=0` y no puedes avanzar.

## 📝 A tener en cuenta

1. Implementa una función que retorne `true/false`.
2. **Complejidad esperada:** tiempo `O(n)` y espacio `O(1)`.
3. Considera casos borde: `nums=[0]`, ceros intermedios, valores grandes, arreglo largo.
