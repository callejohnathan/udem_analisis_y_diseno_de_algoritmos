Se dice que Vasya logró escribir la palabra `p` si es posible borrar algunas letras de `s` (posiblemente ninguna), **sin cambiar el orden de las letras restantes**, de modo que lo que queda sea exactamente `p`.

Por ejemplo, si `p = "hello"`, entonces `s = "ahhellllloou"` cuenta como saludo correcto, pero `s = "hlelo"` no, porque las letras de `p` no aparecen en el orden correcto.

### Objetivo

Dados `s` y `p`, determinar si Vasya logró escribir la palabra `p`.

### Entrada

* Un string `s` (1 ≤ `len(s)` ≤ 10^5), compuesto solo por letras minúsculas.
* Un string `p` (1 ≤ `len(p)` ≤ 10^5), compuesto solo por letras minúsculas.

### Salida

* `"SI"` si es posible obtener `p` borrando letras de `s`, `"NO"` en caso contrario.

### Ejemplos

* `s = "ahhellllloou"`, `p = "hello"` → `"SI"`
* `s = "hlelo"`, `p = "hello"` → `"NO"`

* `s = "bbaannaannaa"`, `p = "banana"` → `"SI"`
* `s = "aaab"`, `p = "aaaa"` → `"NO"`
* `s = "udem"`, `p = "udem"` → `"SI"`
* `s = "programacion"`, `p = "gram"` → `"SI"`
