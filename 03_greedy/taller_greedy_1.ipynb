## Punto 1 — Selección óptima de carga

Una empresa de logística debe transportar materiales en un vehículo cuya capacidad máxima de carga es limitada.

Cada material tiene dos características:

* **Valor**: representa la utilidad o importancia del material.
* **Peso**: representa cuánto espacio o carga ocupa en el vehículo.

El vehículo puede transportar como máximo **W unidades de peso**.

En algunos casos, **es posible transportar solo una fracción de un material** si no cabe completamente en el vehículo.

### Problema

Diseñe un algoritmo que determine **el valor máximo total que se puede transportar** sin superar la capacidad del vehículo.

### Entrada

* Una lista de materiales representados como tuplas `(valor, peso)`.
* Un entero `W` que representa la capacidad máxima del vehículo.

Ejemplo de entrada:

```python
materiales = [(60,10), (100,20), (120,30)]
W = 50
```

### Salida

Un número que represente el **valor máximo total** que se puede transportar.

Para el ejemplo anterior, la salida sería:

```
240
```

### Restricciones

* Todos los valores y pesos son números positivos.
* La capacidad `W` es un entero positivo.
* Puede transportarse **solo una fracción de un material** si el peso restante no permite transportarlo completo.

# Punto 2 — Subarreglo más largo con a lo sumo K valores distintos

Dado un arreglo de enteros `A` y un entero `K`, determine la **longitud máxima de un subarreglo continuo** que contenga **a lo sumo `K` valores distintos**.

Un **subarreglo continuo** es una secuencia de elementos consecutivos dentro del arreglo original.

### Entrada

* Un arreglo de enteros `A`.
* Un entero `K`.

Ejemplo:

```python
A = [1,2,1,2,3]
K = 2
```

### Salida

Un entero que represente la **longitud máxima** de un subarreglo continuo que contenga como máximo `K` valores distintos.

Para el ejemplo anterior, la salida sería:

```
4
```

### Explicación del ejemplo

El subarreglo:

```
[1,2,1,2]
```

contiene únicamente **dos valores distintos** (`1` y `2`) y tiene longitud `4`, que es la mayor posible cumpliendo la condición.
