## Enunciado 1 — Boletos fuertemente ordenados *(adaptado de Codeforces 160B "Unlucky Ticket")*

En la empresa de transporte municipal, cada boleto de bus está compuesto por `2n` dígitos. Tradicionalmente, un boleto se considera *de la suerte* si la suma de los dígitos de su primera mitad es igual a la suma de los dígitos de su segunda mitad.

El supervisor de la empresa quiere detectar, en cambio, los boletos **fuertemente ordenados**: un boleto es fuertemente ordenado si es posible emparejar cada uno de los `n` dígitos de la primera mitad con exactamente un dígito de la segunda mitad (es decir, una correspondencia uno a uno entre ambas mitades) de forma que se cumpla una sola de estas dos condiciones, para todos los pares a la vez:

* cada dígito de la primera mitad es estrictamente menor que su pareja en la segunda mitad, o
* cada dígito de la primera mitad es estrictamente mayor que su pareja en la segunda mitad.

No está permitido mezclar ambas relaciones: si para algún par se cumple "menor" y para otro par se cumple "mayor", el boleto no es fuertemente ordenado.

### Objetivo

Dado un boleto de `2n` dígitos, determinar si es fuertemente ordenado.

### Entrada

* Un entero `n` (1 ≤ n ≤ 100).
* Un string `boleto` de `2n` caracteres, cada uno un dígito `'0'`-`'9'`.

### Salida

* `"SI"` si existe una correspondencia que cumpla el criterio, `"NO"` en caso contrario.

### Ejemplos

* `n = 2`, `boleto = "2421"` → `"SI"` (correspondencia posible: 2 > 1 y 4 > 2)
* `n = 2`, `boleto = "0135"` → `"SI"` (correspondencia posible: 0 < 3 y 1 < 5)
* `n = 2`, `boleto = "3754"` → `"NO"` (no existe ninguna correspondencia uniforme)
