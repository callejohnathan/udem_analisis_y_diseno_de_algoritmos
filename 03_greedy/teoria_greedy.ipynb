---

## 🚩 Idea central y definición

Un **algoritmo voraz** construye una solución **paso a paso**, eligiendo en cada paso la opción localmente óptima (la “mejor” disponible según un criterio) con la **esperanza** de alcanzar un óptimo global.

* 🔑 **Regla voraz (greedy-choice rule):** criterio local para escoger el siguiente elemento/acción.
* 🧱 **Estructura de optimalidad (optimal substructure):** el problema óptimo contiene soluciones óptimas de subproblemas.
* ✅ **Intercambiabilidad/Matroides:** en problemas con ciertas propiedades, cualquier solución se puede transformar a una solución voraz sin perder optimalidad.

---

## 🧪 ¿Cuándo *sí* funciona lo voraz?

Para que un algoritmo greedy sea correcto, normalmente necesitamos **dos** ingredientes:

1. **Subestructura óptima** (como en DP).
2. **Propiedad de elección voraz:** podemos **forzar** que exista un óptimo que comience con la elección local óptima.

### Tres marcos de corrección (muy usados)

* 🧩 **Argumento de intercambio (exchange argument):** si tu solución óptima no sigue mi elección greedy inicial, puedo **intercambiar** decisiones para obtener otra solución óptima que sí la siga, sin empeorar el valor.
* ✂️ **Propiedad de corte (cut property):** en problemas de grafos (MST), la arista de mínimo peso que cruza cualquier corte seguro es parte de **alguna** MST (Kruskal/Prim).
* 🏛️ **Matroides:** si las soluciones factibles forman un matroide y el peso es aditivo, **ordenar por peso y tomar lo mejor que no viole factibilidad** es óptimo (teorema de matroides).

---

## 🎯 Patrón de diseño “Greedy-Recipe”

1. **Modela** el problema:

   * ¿Maximizas/minimizas? ¿Hay restricciones (capacidad, compatibilidad, independencia)?
2. **Propón un criterio local**:

   * “tomar el más barato”, “la mayor ganancia por unidad”, “el más temprano en terminar”, etc.
3. **Verifica factibilidad incremental**:

   * ¿Añadir tu elección mantiene las restricciones?
4. **Intenta una prueba**:

   * Intercambio, corte o matroide. Si falla, revisa el criterio.
5. **Analiza complejidad**:

   * Típico: ordenar `O(n log n)` + inserciones/chequeos `O(n α(n))` o `O(n log n)` con estructuras.

---

## 🧰 Problemas canónicos (y por qué Greedy es correcto)

### 4.1 Interval Scheduling (máx. número de actividades compatibles)

* **Regla voraz:** ordenar por *tiempo de finalización más temprano* y tomar actividades no solapadas.
* **Prueba:** argumento de intercambio; si un óptimo elige otra actividad que acaba más tarde, podemos **intercambiarla** por la que acaba antes, sin reducir el # de actividades.
* ⏱️ Complejidad: `O(n log n)` por el sort.


* Paso a paso:

  1. Ordenar por fin minimiza el “bloqueo” futuro.
  2. Cada elección mantiene factibilidad.
  3. Intercambio demuestra que no perdemos optimalidad.

---

### 4.2 Huffman Coding (codificación óptima por frecuencias)

* **Regla voraz:** combinar repetidamente las **dos** frecuencias más pequeñas.
* **Prueba:** por **intercambio** y optimalidad inductiva: en algún óptimo, las dos hojas de profundidad máxima son precisamente las menos frecuentes unidas primero.
* ⏱️ `O(n log n)` con heap.

---

### 4.3 MST (Kruskal/Prim)

* **Kruskal:** ordenar aristas por peso y tomar la más barata que no forme ciclo (DSU).
* **Prim:** crecer desde un nodo eligiendo arista mínima que conecta el árbol parcial.
* **Prueba:** **propiedad de corte**.
* ⏱️ `O(m log n)` típico.

---

### 4.4 Dijkstra (pesos no negativos)

* **Regla voraz:** siempre expandir el vértice con **distancia mínima provisional**.
* **Prueba:** si existiera un camino más corto no descubierto, violaría la elección del mínimo no negativo.
* ⚠️ **Falla** si hay pesos negativos ⇒ no es correcto.

---

### 4.5 Fractional Knapsack (mochila fraccionaria)

* **Regla:** tomar por **mayor valor/peso** permitiendo fracciones.
* **Prueba:** por intercambio; cualquier desviación puede mejorarse moviendo valor/peso más alto.
* 🚫 **No** aplica a 0/1 Knapsack (contraejemplo clásico).

---

## 🚫 Contraejemplos (dónde **no** funciona)

* **0/1 Knapsack:** elegir por mayor valor/peso puede ser subóptimo (ej.: un item grande de ratio alto impide meter dos medianos mejores).
* **Cambio de monedas (mín # de monedas):** greedy funciona en sistemas **canónicos** (p. ej., monedas {1,5,10,25} en USD), pero falla en sistemas arbitrarios (p. ej., {1,3,4} para 6).
* **Caminos mínimos con pesos negativos:** Dijkstra no sirve; usar Bellman-Ford.
* **Interval Partitioning (min # de salas):** ordenar por fin no basta; la solución óptima usa un heap por **inicio** y asignación a recursos (otro tipo de greedy).

---

## 🧩 Greedy vs Programación Dinámica

| Aspecto     | Greedy                                                             | DP                                                  |
| ----------- | ------------------------------------------------------------------ | --------------------------------------------------- |
| Elección    | Localmente óptima                                                  | Explora subproblemas (tabla/memo)                   |
| Correctitud | Requiere prueba de greedy-choice                                   | Subestructura óptima + recurrencia                  |
| Complejidad | Suele ser `O(n log n)` o `O(n)`                                    | Suele ser `O(n^2)` o más                            |
| Uso típico  | MST, Dijkstra (no negativos), intervalos, Huffman, fracc. knapsack | 0/1 knapsack, edición de strings, conteos complejos |

**Heurística:** si puedes demostrar **intercambio/corte** o identificar un **matroide**, Greedy es fuerte. Si no, intenta DP.

## 🧪 Plantillas (pseudocódigo) reutilizables

### 7.1 “Ordenar y escoger si no viola”

```
sort(S, by = clave)
sol = ∅
for x in S:
    if factible(sol ∪ {x}):
        sol ← sol ∪ {x}
return sol
```

* Usos: interval scheduling, selección de tareas compatibles, selección de aristas sin ciclo (Kruskal con DSU en factible).

### 7.2 “Heap y combinar mínimos” (Huffman)

```
H ← heap(S)  # frecuencias
while |H| > 1:
    a ← extraer_min(H)
    b ← extraer_min(H)
    insertar(H, combinar(a,b))
return arbol
```

### 7.3 “Expandir frontera mínima” (Dijkstra/Prim)

```
dist[s]=0; push (0,s) en heap
while heap:
    d,u ← pop_min()
    if d != dist[u]: continue
    for (u,v,w):
        if dist[v] > dist[u]+w:
            dist[v] = dist[u]+w
            push(dist[v], v)
```


## 🧭 Estrategias para encontrar la **regla voraz**

1. **Ordenar por “terminar rápido”** (fin temprano, duración corta, fecha límite).
2. **Ordenar por densidad/ratio** (valor/peso) — cuidado: no siempre es correcto (0/1 knapsack).
3. **“El más barato que no rompa”** (MST: peso mínimo sin ciclo).
4. **“Combinar mínimos”** (Huffman).
5. **“Expandir el mínimo actual”** (Dijkstra, Prim): estructura incremental.

Si **no puedes** justificar con intercambio/corte, sospecha que necesitas DP.

---

## 🧨 Errores comunes

* Usar Dijkstra con pesos negativos.
* Aplicar razón valor/peso a 0/1 knapsack.
* Confundir “más actividades” con “mayor tiempo total” en scheduling.
* Ignorar la **factibilidad incremental** (e.g., capacidad, incompatibilidades).

---

## 17) 📌 Conclusiones operativas

* Greedy es **potente** cuando puedes justificar **elección local segura**.
* Si dudas, intenta formalizar un **intercambio** o una **propiedad de corte**.
* Sin esa justificación, considera **DP**.
* Las estructuras (heap, DSU) suelen ser clave para eficiencia.

# ✅ Checklist rápido: ¿Aplica Greedy aquí?

> Objetivo: decidir **si** intentar un algoritmo voraz y **cómo** justificar su correctitud. Incluye banderas 🟢/🔴, plantillas de prueba y patrones típicos.

---

## 1) Diagnóstico express (sí/no)

* 🟢 **Objetivo aditivo o “suma de costos/beneficios”?**
  (p. ej., peso total, número de tareas, longitud total) → Greedy suele encajar.

* 🟢 **Feasibilidad hereditaria** (si una solución es factible, cualquier **subconjunto** también lo es).
  ⇒ Huele a **matroide** ➜ Greedy fuerte (ej.: selección de intervalos, MST).

* 🟢 **Elección local “segura”** parece no arruinar el futuro (intuición de *staying ahead*).
  (ej.: actividad que termina más temprano, arista más barata sin ciclo).

* 🟢 **Puedes ordenar y “aceptar si no viola”** sin backtracking.
  (sort + scan + check factibilidad incremental).

* 🟡 **Densidad/ratio** (valor/peso) suena natural.
  ✅ correcto en *fraccional*; ⚠️ sospechoso en 0/1.

* 🔴 **Restricciones acopladas/globales** (capacidad exacta, conteos exactos, incompatibilidades complejas).

* 🔴 **Efectos a largo plazo** (decisiones tempranas bloquean combinaciones óptimas futuras).

* 🔴 **Pesos negativos o no monótonos** (Dijkstra falla).

* 🔴 **Óptimo depende de combinaciones específicas** (0/1 knapsack, intervalos con pesos → pide DP).

👉 Regla práctica: si marcas ≥3 🟢 (incluyendo “hereditaria” o “aceptar si no viola”), **intenta Greedy** y pasa a la prueba de correctitud.

---

## 2) Plantillas de **prueba de correctitud**

### 🧩 A) **Argumento de intercambio (exchange argument)**

1. Sea `OPT` un óptimo que **no** sigue tu primera elección greedy.
2. Identifica la primera posición donde difieren.
3. **Intercambia** el elemento de `OPT` por el elegido por Greedy, manteniendo factibilidad y **sin empeorar** el objetivo.
4. Repite hasta transformar `OPT` en la solución greedy ⇒ Greedy es óptimo.

> Típicos: selección de intervalos por fin temprano, SPT (minimizar ∑ tiempos de finalización), Huffman (combinar dos mínimos).

### ✂️ B) **Propiedad de corte (cut property)**

* En grafos: la arista mínima que cruza **cualquier** corte es **segura**.
  ⇒ Justifica **Kruskal/Prim** para MST.

### 🛤️ C) **Staying ahead / Prefijo-óptimo**

* Tras cada paso, la solución greedy **no va peor** que cualquier otra en el mismo prefijo.
  ⇒ “Siempre voy por delante”. Funciona en varios scheduling.

### 🧱 D) **Matroide**

* Verifica **hereditariedad** + **intercambio** de independientes.
* Con pesos aditivos: **ordenar por peso** y añadir “si no viola” es óptimo.

> Si ninguna plantilla cuadra, Greedy probablemente **no** aplica como exacto (considera DP o un aproximado).

---

## 3) Patrones donde Greedy suele funcionar

* ⏰ **Scheduling sin pesos**: máximo número de actividades compatibles → **fin más temprano**.
* 🧮 **SPT**: ordenar por duración ascendente para minimizar ∑ Cᵢ.
* 🌲 **MST**: Kruskal/Prim (cut property + DSU/heap).
* 🗺️ **Dijkstra** con pesos **no negativos**.
* 🧵 **Huffman**: combinar dos mínimas frecuencias.
* 🧳 **Mochila fraccional**: por **valor/peso** (no 0/1).
* 🧱 **Interval Partitioning** (min # salas): heap por tiempos de fin.

---

## 4) Banderas rojas (probablemente **no** Greedy exacto)

* 🎒 **0/1 Knapsack** (ratio falla).
* 🪙 **Cambio de monedas** con sistema no canónico (p. ej., {1,3,4} para 6).
* 🧩 **Intervalos con pesos** (Weighted Interval Scheduling) → DP con búsqueda binaria.
* ➖ **Caminos con pesos negativos** → Bellman-Ford.
* 🔗 **Dependencias/precedencias complejas** que rompen hereditariedad.

---

## 5) Heurísticas para **descubrir** la regla voraz

* 📐 “**Terminar antes**” (minimiza bloqueo futuro): ordenar por fin.
* 🧷 “**El más corto primero**” (reduce coste acumulado): SPT.
* 🧲 “**El más barato que no rompa**” (factibilidad incremental): Kruskal.
* ⚖️ “**Mejor densidad**” (si fraccional o submodular-approx).
* 🧰 “**Combinar mínimos**” cuando el coste de fusión crece con los tamaños: Huffman.
* 🔭 **Expandir frontera mínima** con invariante de optimalidad local: Dijkstra/Prim.

> Si hay **empates**, **define** la regla de desempate; puede ser crucial para mantener la prueba.

---

## 6) Lista de **comprobación** antes de comprometerte

* [ ] **Definí** una elección local clara (criterio exacto + desempates).
* [ ] **Demostré** una de: intercambio / corte / staying-ahead / matroide.
* [ ] **Factibilidad incremental** garantizada (no viola restricciones al añadir).
* [ ] **Invariante** explícito que se mantiene paso a paso.
* [ ] **Contraejemplos pequeños** no rompen la regla (intenté construirlos).
* [ ] **Complejidad**: típicamente `O(sort) + O(n log n)` o `O(n)`; memoria `O(1)`/`O(n)` razonable.

---

## 7) Si no es exacto, ¿sirve como **aproximación**?

* 📚 **Set Cover**: greedy por “mejor cobertura/ costo” da `O(log n)`-aprox.
* 📉 **Submodular monotona + restricción de cardinalidad**: greedy (1−1/e)-aprox.
* 🧵 **Empaquetado/scheduling**: varias reglas greedy dan 2-aprox o similares.

> Tip: detecta **submodularidad** (rendimientos decrecientes) para justificar greedy de aproximación.

---

## 8) Mini-flowchart mental (30s)

1. ¿Puedo **ordenar** elementos y **aceptar si no viola**? → prueba intercambio/matroide.
2. ¿Es de **grafos**? ¿Corte seguro? → MST. ¿Pesos ≥0? → Dijkstra.
3. ¿**Fusionar/combinar** repetidamente lo más pequeño? → Huffman-like.
4. ¿**Fraccional**? → ratio; ¿0/1? → sospecha DP.
5. ¿Falla algún contraejemplo simple? → abandona greedy exacto, prueba DP/approx.

---

## 9) Trucos prácticos de implementación

* 🧾 **Preprocesa** patrones (p. ej., colapsar `**`→`*`) para simplificar.
* 🧺 **Estructuras**:

  * Sort + scan;
  * **Heap** (mín./máx.) para “mejor siguiente”;
  * **DSU** para “no formar ciclo” o “slot más tarde posible”;
  * **Dos punteros** cuando hay **monotonicidad** en extremos.
* 🧪 **Tests dirigidos**: construye instancias donde tu elección local podría ser riesgosa; si sobrevives, fortalece la prueba formal.

---

### TL;DR

* **Greedy** es ideal cuando hay **hereditariedad + intercambio** (o **corte**), objetivo **aditivo** y factibilidad **incremental**.
* Si tus decisiones tempranas pueden bloquear combinaciones futuras valiosas → **probablemente DP**.
* Sin prueba (intercambio/corte/staying-ahead/matroide), **no** aceptes Greedy como exacto.


# 🧠 🎯 Intuición para detectar Greedy Choice

Hay tres enfoques mentales que puedes enseñarles:

---

## 1️⃣ Pregunta clave:

> ¿Si tomo esta decisión ahora, podría estar cerrando una mejor solución futura?

Si la respuesta es **sí**, sospecha que greedy no sirve.
Si la respuesta es **no**, puede ser candidato.

---

## 2️⃣ Pensamiento de “intercambio” (Exchange Argument)

Este es el método más poderoso.

La idea intuitiva es:

> Si existe una solución óptima que NO tomó mi decisión greedy,
> ¿puedo transformarla en otra solución óptima que SÍ la tome
> sin empeorar el resultado?

Si puedes hacer ese “intercambio” sin perder optimalidad →
la greedy choice property se cumple.

### Intuición:

* Supón que tu solución óptima eligió A
* Greedy eligió B (porque es mejor localmente)
* Si reemplazar A por B no daña la solución
* Entonces B puede estar en alguna solución óptima

Eso es todo.

---

## 3️⃣ Observa la estructura del problema

Greedy suele funcionar cuando:

✔️ Las decisiones son independientes
✔️ No hay efectos secundarios complejos
✔️ El orden importa más que la combinación
✔️ Elegir algo “mejor ahora” no bloquea combinaciones futuras críticas

Y suele fallar cuando:

❌ Hay combinaciones que superan decisiones individuales
❌ El valor depende de cómo se agrupan elementos
❌ Hay compensaciones futuras no visibles localmente

# 🧩 Regla práctica

Si el problema parece:

* "ordenar + seleccionar"
* "escoger el menor / mayor / primero"
* "maximizar cantidad bajo restricción lineal"

Puede ser greedy.

Si el problema parece:

* "combinaciones"
* "subconjuntos"
* "dependencias entre decisiones"

Cuidado.

Hazte siempre estas tres preguntas:

1. ¿Qué significa “mejor” localmente?
2. ¿Esa elección podría bloquear algo mejor después?
3. ¿Puedo demostrar que cualquier solución óptima puede modificarse para incluir mi elección?

Si no pueden responder la tercera → no hay garantía.

# 🎯 1️⃣ Definición formal — Subestructura Óptima

Un problema tiene **subestructura óptima** si:

> Una solución óptima global contiene soluciones óptimas a sus subproblemas.

Formalmente:

Si ( S^* ) es solución óptima del problema ( P ),
y ( P' ) es un subproblema de ( P ),
entonces la parte de ( S^* ) que resuelve ( P' ) también es óptima para ( P' ).

---

## 🧠 Traducción intuitiva

Si resolviste el problema grande de la mejor manera posible,
entonces necesariamente resolviste cada parte pequeña de la mejor manera posible.

Si en alguna parte pequeña no fuera óptimo,
podrías mejorar el todo.

---

# ✅ 2️⃣ Ejemplo donde SÍ aplica

## 🛣️ Camino más corto (Dijkstra)

Si el camino más corto de A → D es:

A → B → C → D

Entonces:

* El camino más corto de A → C
  también es óptimo.

Si no lo fuera,
podrías reemplazarlo por uno mejor
y mejorar el camino completo.

👉 Claramente hay subestructura óptima.

---

## 🗓️ Interval Scheduling

Si tienes el conjunto máximo de intervalos compatibles:

* Después de elegir uno,
* El resto del problema es otro intervalo scheduling más pequeño.

La solución óptima del problema completo
contiene una solución óptima del subproblema restante.

---

# ❌ 3️⃣ Ejemplo donde NO aplica

Aquí es donde se vuelve interesante.

## 🧩 Problemas donde el contexto cambia el valor

Imagina un problema donde:

* Elegir algo cambia las reglas futuras.
* El valor depende del historial completo.

Ejemplo conceptual:

Supón que el valor de un elemento depende de si ya elegiste otro.

Si el beneficio de un subproblema depende de decisiones externas,
la solución óptima del subproblema aislado
puede no ser parte de la solución óptima global.

---

## 🎒 Mochila con restricciones adicionales

En 0/1 knapsack clásico:

✔️ Sí hay subestructura óptima
(pero no greedy choice)

Pero si agregas restricciones como:

* Bonificación por pares específicos de objetos
* Penalización por combinaciones

Entonces:

El subproblema ya no es independiente.
Se rompe la subestructura óptima.

---

# 🔎 Punto pedagógico clave

⚠️ Subestructura óptima ≠ Greedy choice.


Ejemplo importante:

| Problema           | Subestructura óptima | Greedy choice |
| ------------------ | -------------------- | ------------- |
| Dijkstra           | ✔️                   | ✔️            |
| Knapsack 0/1       | ✔️                   | ❌             |
| Cambio no canónico | ✔️                   | ❌             |

Esto es oro didáctico.

---

# 🧠 Cómo explicarlo intuitivamente

Hazte esta pregunta:

> Si soluciono perfectamente la parte pequeña,
> ¿eso garantiza que estoy contribuyendo a la mejor solución global?

Si la respuesta es sí → hay subestructura óptima.

Si la respuesta es no → cuidado.




---

# ❌ Ejemplo claro donde NO hay subestructura óptima

## 🎯 Problema: Camino más corto SIN permitir repetir colores consecutivos

Supón un grafo donde:

* Cada arista tiene un **peso**
* Cada arista tiene un **color** (rojo o azul)
* Restricción: **no puedes usar dos aristas del mismo color seguidas**

---

### 🔎 ¿Qué cambia?

El costo total ya no depende solo de la suma de pesos.
También depende del **color de la arista anterior**.

El estado ya no es simplemente “estoy en el nodo X”.

Ahora el estado es:

[
(nodo, color_previo)
]

---

## 🧠 Contraejemplo conceptual

Imagina que quieres ir de A → D.

El camino más corto a C (sin restricciones) es:

A → B → C  (ambas rojas)

Pero si llegas a C con roja,
no puedes tomar una roja después.

Tal vez el camino óptimo global requiere:

A → E → C (azul, luego roja)

Ese camino hacia C es más largo,
pero permite completar el camino a D de manera óptima.

---

### 🔥 Lo importante

El camino más corto a C **aisladamente**
NO es necesariamente parte del camino más corto a D.

Entonces:

La solución óptima del subproblema
("mejor camino hasta C")

NO está contenida necesariamente
en la solución óptima del problema completo.

---

# 🎯 Conclusión clave

No hay subestructura óptima porque:

> La mejor solución parcial depende del contexto futuro.

El subproblema no es independiente.

---

# 🧩 Forma simple de explicarlo

A tener en cuenta

Si para resolver el subproblema necesito saber algo del pasado
(o algo del futuro cambia su valor),

entonces el subproblema no es autónomo.

Y si no es autónomo →
no hay subestructura óptima clásica.


# ❌ Ejemplo: Juego de selección con bonificación por combinación

## 🎯 Problema

Tienes tres objetos:

| Objeto | Valor |
| ------ | ----- |
| A      | 10    |
| B      | 9     |
| C      | 1     |

Regla especial:

* Si eliges **B y C juntos**, recibes una bonificación de +15.

Objetivo: maximizar el valor total.

---

## 🔎 Evaluemos combinaciones

* A → 10
* B → 9
* C → 1
* A + B → 19
* A + C → 11
* B + C → 9 + 1 + 15 = **25**

La mejor solución es:

👉 B + C = 25

---

# 🎯 Ahora viene lo importante

Miremos el subproblema:

> ¿Cuál es el mejor objeto individualmente?

Claramente es A (10).

Pero en la solución óptima global (B + C),
NO aparece A.

---

## 🔥 ¿Dónde se rompe la subestructura óptima?

La solución óptima del problema completo:

B + C

NO contiene la solución óptima del subproblema
“elegir el mejor objeto individual”.

El valor de elegir C depende de si también eliges B.

Los subproblemas no son independientes.

---

# 🧠 Intuición

Cuando el valor de una decisión depende de otra decisión:

👉 No hay subestructura óptima clásica.

Porque el subproblema no se puede resolver aislado.

---

# 🎓 Frase poderosa para cerrar

Si el beneficio de una parte depende del contexto completo,
no puedes asumir que resolver la parte óptimamente
te ayudará a resolver el todo óptimamente.

---
